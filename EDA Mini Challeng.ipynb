# Topic 9 – Outlier Detection

Now we're going to find unusual values in numerical columns.

For this challenge, practice the IQR method.

Your task

For the Customer_Age column, find:

Q1
Q3
IQR
Lower limit
Upper limit
Values outside those limits

```python
Q1 = df["Customer_Age"].quantile(0.25)
Q3 = df["Customer_Age"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Customer_Age"] < lower_limit) |
    (df["Customer_Age"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
print(outliers)
```

```python
Let's solve each one.

Chart 1 — Category Comparison

We want to know:

How many orders belong to each category?

First, value_counts() counts each category:

df["Category"].value_counts()

Then we want a bar chart, because we're comparing categories.

kind="bar"

So:

import matplotlib.pyplot as plt


df["Category"].value_counts().plot(kind="bar")


plt.xlabel("Category")
plt.ylabel("Number of Orders")
plt.title("Orders by Category")
plt.show()
Remember

Categorical comparison → Bar chart

Chart 2 — Numerical Relationship

We want to see:

Does Quantity have a relationship with Unit_Price?

Both are numerical variables.

The appropriate chart is a scatter plot.

df.plot(
    x="Quantity",
    y="Unit_Price",
    kind="scatter"
)


plt.xlabel("Quantity")
plt.ylabel("Unit Price")
plt.title("Quantity vs Unit Price")
plt.show()
Remember

Numerical + Numerical → Scatter plot

Chart 3 — Outlier Visualization

We want to find unusual values in Unit_Price.

The appropriate chart is a box plot.

df["Unit_Price"].plot(kind="box")


plt.ylabel("Unit Price")
plt.title("Unit Price Outlier Analysis")
plt.show()
Remember this chart-selection rule
Categorical comparison     → Bar chart
Numerical distribution     → Histogram
Numerical + Numerical      → Scatter plot
Outlier detection          → Box plot
Correlation matrix         → Heatmap

You've now covered the main technical EDA steps. ✅

T
```
