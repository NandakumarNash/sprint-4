### Numerical vs Numerical

### Definition:
Numerical vs Numerical analysis studies the relationship between two numerical variables.

Both variables contain numeric values.

### Example:
Age vs Salary

- Age     Salary
- 22    -  25,000
- 25    -  32,000
- 30    -  42,000
- 35    -  55,000
- 40    -  65,000

### Technique:
Scatter Plot

### Observation:
As Age increases, Salary generally increases.

### Relationship:
There is a positive relationship between Age and Salary.

### Real-time Example:
In a company, employees with more experience may generally earn higher salaries.

### AI/ML Use:
Bivariate analysis helps identify whether one feature may be useful for predicting another variable or the target.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample data
df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Salary": [25000, 32000, 42000, 55000, 65000]
})

# Scatter plot
sns.scatterplot(x="Age", y="Salary", data=df)
plt.show()
```

### Numerical vs Categorical

Definition:
Numerical vs Categorical analysis studies the relationship between one numerical variable and one categorical variable.

### Example:
Department vs Salary

Department → Categorical
Salary     → Numerical

### Data:
Department    Salary
- IT         -   60000
- IT         -   65000
- HR         -  40000
- HR         -  45000
- Finance    -   50000
- Finance    -   55000

### Technique:
Box Plot

### Observation:
Salary values are different across departments.

### Relationship:
IT has higher salary values than HR and Finance.

### Real-time Example:
In a company, different departments may have different salary ranges.

### AI/ML Use:
This helps us understand whether a categorical feature such as Department has a relationship with a numerical target such as Salary.

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample data
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Finance", "Finance"],
    "Salary": [60000, 65000, 40000, 45000, 50000, 55000]
})

# Box plot
sns.boxplot(x="Department", y="Salary", data=df)
plt.show()
```

## 3. Categorical vs Categorical

### Definition
Categorical vs Categorical analysis studies the relationship between two categorical variables.

### Example
Product Category vs Sales Level

- Product Category → Categorical
- Sales Level → Categorical

### Data

| Product Category | Sales Level |
|---|---|
| Electronics | High |
| Electronics | High |
| Clothing | Medium |
| Clothing | Low |
| Grocery | Low |
| Grocery | Medium |

### Technique
Grouped Bar Chart or Cross Tabulation

### Observation

Electronics has mostly High sales, while Grocery has Low and Medium sales.

### Relationship

Product Category and Sales Level appear to have a relationship because the distribution of sales levels differs across product categories.

### Real-Time Example

A retail company can analyze whether certain product categories are more likely to have High, Medium, or Low sales.

### AI/ML Use

This analysis helps identify relationships between categorical features and can help with feature selection and understanding customer or business patterns.



```python
### Python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample data
df = pd.DataFrame({
    "Product_Category": [
        "Electronics", "Electronics",
        "Clothing", "Clothing",
        "Grocery", "Grocery"
    ],
    "Sales_Level": [
        "High", "High",
        "Medium", "Low",
        "Low", "Medium"
    ]
})

# Grouped bar chart
sns.countplot(
    x="Product_Category",
    hue="Sales_Level",
    data=df
)

plt.show()
```

## 6. Violin Plot

### Definition
A Violin Plot is used to compare the distribution of a numerical variable across different categories.

It combines the information of a box plot with the **density/shape of the distribution**.

### Example
Department vs Salary

- Department → Categorical
- Salary → Numerical


```python

### Python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["IT", "IT", "IT", "IT", "HR", "HR", "HR", "HR"],
    "Salary": [50000, 55000, 60000, 65000, 30000, 35000, 40000, 45000]
})

sns.violinplot(x="Department", y="Salary", data=df)

plt.title("Department vs Salary")
plt.show()
```

## 8. Cross Tabulation

### Definition
Cross Tabulation (Crosstab) is a table that shows the frequency or count of combinations between two categorical variables.

### Example
Product Category vs Sales Level

- Product Category → Categorical
- Sales Level → Categorical


```python

### Python
import pandas as pd

df = pd.DataFrame({
    "Product_Category": [
        "Electronics", "Electronics", "Electronics",
        "Clothing", "Clothing",
        "Grocery", "Grocery", "Grocery"
    ],
    "Sales_Level": [
        "High", "High", "Medium",
        "Medium", "Low",
        "Low", "Medium", "Low"
    ]
})

# Cross Tabulation
pd.crosstab(
    df["Product_Category"],
    df["Sales_Level"]
)
```
