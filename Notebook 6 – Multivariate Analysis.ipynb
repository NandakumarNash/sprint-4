# Correlation Matrix

## Definition

A **Correlation Matrix** is a table that shows the correlation between every pair of numerical features in a dataset.

The correlation value ranges from **-1 to +1**.

* **+1** → Perfect positive relationship
* **+0.7** → Strong positive relationship
* **0** → No linear relationship
* **-0.7** → Strong negative relationship
* **-1** → Perfect negative relationship

## Easy Real-World Example – House Prices 🏠

Suppose a house dataset contains:

* `Area`
* `Bedrooms`
* `Age`
* `Price`

We want to understand how these numerical features are related.

## Interpretation

* `Area ↔ Price = 0.98` → Very strong positive correlation. Larger houses tend to have higher prices.
* `Age ↔ Price = -0.92` → Strong negative correlation. Older houses tend to have lower prices.
* `Area ↔ Bedrooms = 0.85` → Strong positive correlation. Larger houses tend to have more bedrooms.
* `Age ↔ Bedrooms = -0.75` → Strong negative correlation.

## Why do we use it in EDA?

A correlation matrix helps us identify:

* Strong correlations
* Weak correlations
* Positive correlations
* Negative correlations
* Possible multicollinearity
* Features that may be useful for further analysis

## Exam Point

**A correlation matrix is a table that shows the correlation between all pairs of numerical variables in a dataset.**


````python
import pandas as pd

data = {
    "Area": [800, 1000, 1200, 1500, 1800],
    "Bedrooms": [2, 2, 3, 3, 4],
    "Age": [15, 12, 10, 5, 2],
    "Price": [40, 50, 65, 80, 100]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix)
````

# Heatmap

## Definition

A **Heatmap** is a visualization that represents values using **different colors**.

In Multivariate Analysis, a heatmap is commonly used to visualize a **correlation matrix**, making strong, weak, positive, and negative relationships easier to identify.

## Easy Real-World Example – Weather 🌡️

Imagine we record the temperature of different cities for different days:

```text
             Monday   Tuesday   Wednesday   Thursday
Chennai        32       34          33          35
Bangalore      25       27          26          28
Delhi          30       32          31          33
Mumbai         29       31          30          32
```

Instead of looking at all these numbers, a heatmap uses different colors to make **high and low temperatures immediately visible**.

## What We Can Understand

The heatmap allows us to quickly identify:

* Which city has higher temperatures
* Which city has lower temperatures
* Which days are hotter or cooler
* Similar temperature patterns across cities

For example:

**Chennai** → consistently higher temperatures.

**Bangalore** → consistently lower temperatures.

## Heatmap in Correlation Analysis

In EDA, we often use a heatmap with a **correlation matrix**:

```python
correlation = df.corr()

sns.heatmap(correlation, annot=True)
plt.show()
```

Here, the colors represent **correlation strength** rather than temperature.

## Why do we use it in EDA?

A heatmap helps us quickly see patterns in a large amount of numerical data without reading every value individually.

It is especially useful for:

* Correlation analysis
* Finding strong relationships
* Finding weak relationships
* Identifying positive and negative correlations
* Detecting possible multicollinearity
* Comparing values across multiple categories

## Easy Way to Remember

**Heatmap = Numbers → Colors → Easy Pattern Identification**

## Exam Point

**A heatmap is a graphical representation of numerical values using colors to make patterns, relationships, and differences easy to identify.**


````python


## Python Example

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Monday": [32, 25, 30, 29],
    "Tuesday": [34, 27, 32, 31],
    "Wednesday": [33, 26, 31, 30],
    "Thursday": [35, 28, 33, 32]
}

df = pd.DataFrame(
    data,
    index=["Chennai", "Bangalore", "Delhi", "Mumbai"]
)

sns.heatmap(df, annot=True, cmap="coolwarm")
plt.title("City Temperature Heatmap")
plt.show()

````

# Pair Plot

## Definition

A **Pair Plot** shows the relationship between **every pair of numerical features** in a dataset.

It combines:

* Scatter plots → relationship between two features
* Distribution plots → distribution of individual features

## Easy Real-World Example – Students 🎓

Suppose we have student data:

* `Study_Hours`
* `Attendance`
* `Assignments`
* `Marks`

We want to understand whether students who study more, attend more classes, and complete more assignments tend to get higher marks.

## What Does Pair Plot Show?

For example:

**Study_Hours ↔ Marks**

If the points move upward, students who study more tend to score higher.

**Attendance ↔ Marks**

If the points move upward, higher attendance may be associated with higher marks.

**Assignments ↔ Marks**

If the points move upward, completing more assignments may be associated with higher marks.

The diagonal plots show the **distribution of each individual feature**.

## Why Do We Use It in EDA?

A pair plot helps us quickly identify:

* Positive relationships
* Negative relationships
* Weak relationships
* Strong relationships
* Outliers
* Feature distributions
* Possible patterns between multiple numerical features

## Pair Plot vs Correlation Matrix

**Correlation Matrix** → Gives a numerical correlation value.

**Pair Plot** → Gives a visual picture of the relationships.

Example:

```text
Correlation:
Study_Hours ↔ Marks = 0.95

Pair Plot:
Shows visually that the points move upward.
```

## Easy Way to Remember

**Pair Plot = Many scatter plots + feature distributions**

## Exam Point

**A pair plot is a visualization that shows pairwise relationships and distributions among multiple numerical features in a dataset.**


````python
## Python Example
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7],
    "Attendance": [60, 65, 70, 75, 85, 90],
    "Assignments": [3, 4, 5, 6, 8, 9],
    "Marks": [45, 50, 58, 65, 78, 88]
}

df = pd.DataFrame(data)

sns.pairplot(df)
plt.show()
````

# Multiple Feature Relationships

## Definition

**Multiple Feature Relationships** means analyzing how **three or more features are related to each other at the same time**.

In bivariate analysis, we mainly study **two variables**.

In multivariate analysis, we look at **multiple variables together** to understand a bigger pattern.

## Easy Real-World Example – Car 🚗

Suppose we have a car dataset:

```text
Car_Model | Engine_Size | Weight | Mileage | Price
```

We can study:

* `Engine_Size ↔ Mileage`
* `Weight ↔ Mileage`
* `Engine_Size ↔ Price`
* `Weight ↔ Price`

But we can also look at them **together**.

For example:

> Does a larger engine and heavier car result in lower mileage and a higher price?

This involves multiple features.

## What Can We Identify?

Suppose we observe:

```text
Engine_Size ↑
        ↓
Mileage ↓
        ↓
Price ↑
```

This tells us that:

* Larger engines tend to have **lower mileage**.
* Larger engines tend to have **higher prices**.
* Heavier cars tend to have **lower mileage**.
* Heavier cars may also have **higher prices**.

These relationships become easier to understand when several features are analyzed together.

## Why Do We Use It in EDA?

Multiple feature analysis helps us:

* Understand complex relationships
* Find important feature combinations
* Identify related features
* Discover patterns that may not be obvious from one pair
* Understand which features may influence the target
* Detect possible multicollinearity

## Bivariate vs Multivariate

**Bivariate:**

```text
Engine_Size ↔ Price
```

Only two variables.

**Multivariate:**

```text
Engine_Size
     ↓
Mileage ← Weight
     ↓
   Price
```

Several variables are considered together.

## Exam Point

**Multiple Feature Relationships means analyzing relationships among three or more features simultaneously to discover patterns and dependencies in a dataset.**


````python

## Python Example
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Engine_Size": [1.2, 1.5, 1.8, 2.0, 2.5, 3.0],
    "Weight": [1000, 1100, 1200, 1300, 1500, 1700],
    "Mileage": [22, 20, 18, 16, 13, 10],
    "Price": [6, 8, 10, 13, 18, 25]
}

df = pd.DataFrame(data)

sns.pairplot(df)
plt.show()

````

# Feature Interactions

## Definition

**Feature Interaction** occurs when the effect of **one feature on the target changes depending on the value of another feature**.

In simple words:

> **Two features work together and their combination affects the outcome.**

## Easy Real-World Example – Employee Salary 💼

Suppose we have:

```text
Experience | Education | Salary
```

Consider two employees:

```text
Employee A:
Experience = 5 years
Education = Bachelor's
Salary = ₹50,000

Employee B:
Experience = 5 years
Education = Master's
Salary = ₹70,000
```

Both employees have the **same experience**, but education changes the salary.

Now consider:

```text
Experience + Education → Salary
```

The effect of **Experience** on salary may be different for employees with different education levels.

This is a **feature interaction**.

## Another Easy Example – House Price 🏠

Suppose:

```text
Area | Location | Price
```

A 1,500 sq.ft house may have different prices depending on its location.

```text
1,500 sq.ft + Normal Location → ₹70 lakh

1,500 sq.ft + Prime Location → ₹1.2 crore
```

Here:

```text
Area + Location → Price
```

The effect of **Area** on price depends partly on **Location**.

## Python Example

We can create an interaction feature:


The new feature:

```text
Experience_Education
```

represents the **combined effect of Experience and Education**.

## Why Do We Use Feature Interactions?

Feature interactions help us:

* Understand combined effects of features
* Discover hidden patterns
* Improve machine learning models
* Create useful interaction features
* Understand why a feature may behave differently for different groups

## Simple Example to Remember

```text
Experience alone → Salary

Education alone → Salary

Experience + Education → Salary
```

The last relationship represents a **feature interaction**.

## Important Difference

**Multiple Feature Relationships** → We examine relationships among several features.

**Feature Interaction** → We specifically look for a situation where **the effect of one feature depends on another feature**.

## Exam Point

**Feature interaction occurs when the effect of one feature on the target variable depends on the value of another feature.**


````python

import pandas as pd

data = {
    "Experience": [1, 2, 3, 5, 7],
    "Education_Level": [1, 1, 2, 2, 3],
    "Salary": [30000, 35000, 45000, 60000, 85000]
}

df = pd.DataFrame(data)

df["Experience_Education"] = (
    df["Experience"] * df["Education_Level"]
)

print(df)
````

# Numerical Feature Relationships

## Definition

**Numerical Feature Relationships** means analyzing how **numerical variables are related to each other**.

We look for patterns such as:

* Positive relationship
* Negative relationship
* Strong relationship
* Weak relationship
* No clear relationship

## Easy Real-World Example – Online Shopping 🛒

Suppose an e-commerce dataset contains:

```text
Quantity | Discount | Sales | Profit
```

We can study relationships such as:

```text
Quantity ↔ Sales
Discount ↔ Profit
Sales ↔ Profit
```

### Example

If customers buy more products:

```text
Quantity ↑ → Sales ↑
```

This indicates a **positive relationship**.

If a higher discount reduces profit:

```text
Discount ↑ → Profit ↓
```

This indicates a **negative relationship**.


## What Can We Identify?

### Positive Relationship

```text
Quantity ↑ → Sales ↑
```

More products sold → higher sales.

### Negative Relationship

```text
Discount ↑ → Profit ↓
```

Higher discount → lower profit.

### Weak Relationship

If the points are scattered without a clear pattern, the relationship may be weak.

## Why Do We Use It in EDA?

Numerical feature relationships help us:

* Understand relationships between numerical variables
* Find strong and weak relationships
* Identify positive and negative relationships
* Find useful features for machine learning
* Detect possible multicollinearity
* Discover unusual patterns

## Common Methods

We can study numerical feature relationships using:

```text
Correlation Matrix
Heatmap
Scatter Plot
Pair Plot
Covariance
```

## Simple Example to Remember

```text
Quantity ↑  → Sales ↑
                ↓
          Positive relationship

Discount ↑ → Profit ↓
                ↓
          Negative relationship
```

## Exam Point

**Numerical feature relationships describe how numerical variables are associated with each other using techniques such as correlation, scatter plots, heatmaps, and pair plots.**


````python
## Python Example


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Quantity": [1, 2, 3, 4, 5, 6],
    "Discount": [5, 10, 15, 20, 25, 30],
    "Sales": [100, 190, 300, 390, 480, 570],
    "Profit": [30, 55, 75, 80, 70, 50]
}

df = pd.DataFrame(data)

print(df.corr())

````

````python
## We can also visualize one numerical relationship:


sns.scatterplot(data=df, x="Quantity", y="Sales")
plt.show()
````
