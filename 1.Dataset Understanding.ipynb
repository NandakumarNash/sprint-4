# 1. What is a Dataset?

## Definition
A dataset is a collection of data organized in rows and columns.

- Rows → Individual records
- Columns → Features or information about each record


## Business Example
A company collects customer data such as Age, Salary, Location, and Purchase History.

The company can use this dataset to predict whether a customer will purchase a product.

## AI/ML Use
A Machine Learning model uses the features to learn patterns and predict the target.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Experience": [1, 5, 8, 12],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```


## Explanation
- df → Dataset
- Rows → Individual employees
- Columns → Information about employees
- Age, Salary, Experience, Department → Features
- Purchased → Target variable


# 2. What is a Feature?

## Definition
A feature is an individual column in a dataset that contains information used by a Machine Learning model.

## Business Example
An online shopping company may use:
- Age
- Salary
- Previous Purchases
- Website Visits

These features can help predict whether a customer will purchase a product.

## AI/ML Use
Features are the input variables given to a Machine Learning model for learning and prediction.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Experience": [1, 5, 8, 12],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df[["Age", "Salary", "Experience"]])
```

## Explanation
- Age → Feature
- Salary → Feature
- Experience → Feature
- Features provide input information to the ML model.

# 3. What is a Target?

## Definition
A target is the value that a Machine Learning model tries to predict.

## Business Example
A bank may use:
- Age
- Income
- Credit Score
- Loan Amount

to predict:

- Loan Approved → Yes/No

Here, **Loan Approved** is the target.

## AI/ML Use
The target is the value the ML model learns to predict.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Experience": [1, 5, 8, 12],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

X = df[["Age", "Salary", "Experience"]]
y = df["Purchased"]

print("Features:")
print(X)

print("\nTarget:")
print(y)

```


## Explanation
- X → Features (input)
- y → Target (output)
- Purchased → Target variable
- The model learns from X to predict y.

# 4. Independent Variables

## Definition
Independent variables are the input variables used to predict the target variable.

## Business Example
A bank wants to predict whether a customer will get a loan.

Independent variables:
- Age
- Salary
- Credit Score
- Existing Loan

Target:
- Loan Approved

## AI/ML Use
Independent variables are used as input features to train the Machine Learning model.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Experience": [1, 5, 8, 12],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

X = df[["Age", "Salary", "Experience"]]

print(X)
```

## Explanation
- Age → Independent variable
- Salary → Independent variable
- Experience → Independent variable
- Purchased → Target variable
- Independent variables are the inputs given to the ML model.

# 5. Dependent Variables

## Definition
A dependent variable is the output or result that depends on the independent variables.

## Business Example
A company wants to predict employee salary.

Independent variables:
- Experience
- Education
- Skills
- Job Level

Dependent variable:
- Salary

## AI/ML Use
The dependent variable is the value the Machine Learning model learns to predict from the independent variables.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Experience": [1, 5, 8, 12],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

X = df[["Age", "Salary", "Experience"]]
y = df["Purchased"]

print("Independent Variables:")
print(X)

print("\nDependent Variable:")
print(y)

## Output

Independent Variables:
   Age  Salary  Experience
0   25   30000           1
1   30   50000           5
2   35   70000           8
3   40   90000          12

Dependent Variable:
0     No
1    Yes
2    Yes
3    Yes
Name: Purchased, dtype: object
```


## Explanation
- Independent variables → Inputs (X)
- Dependent variable → Output (y)
- Purchased depends on information such as Age, Salary, and Experience.

# 7. Categorical Features

## Definition
Categorical features contain labels or groups instead of numerical measurements.

## Business Example
An e-commerce company may have:
- Product Category → Electronics, Clothing, Grocery
- Payment Method → UPI, Card, Cash
- Customer Type → New, Returning

These are categorical features.

## AI/ML Use
Categorical features are converted into numerical values using techniques such as:
- One-Hot Encoding
- Label Encoding

This allows ML models to use categorical information.

```python

## Python Example

import pandas as pd

data = {
    "Department": ["IT", "HR", "Finance", "IT"],
    "Gender": ["Male", "Female", "Male", "Female"],
    "Purchased": ["Yes", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print(df)
```


## Explanation
- Department → Categorical feature
- Gender → Categorical feature
- Purchased → Categorical target
- Categories represent groups or labels.

# 8. Discrete Features

## Definition
Discrete features are numerical features that contain countable, whole-number values.

## Business Example
An e-commerce company may track:
- Number of products purchased
- Number of website visits
- Number of orders
- Number of complaints

These are discrete features because they can be counted.

## AI/ML Use
Discrete features can be used as input features to help an ML model identify patterns and make predictions.

```python

## Python Example

import pandas as pd

data = {
    "Number_of_Children": [0, 1, 2, 3],
    "Number_of_Purchases": [2, 5, 8, 10],
    "Age": [25, 30, 35, 40]
}

df = pd.DataFrame(data)

print(df)
```


## Explanation
- Number_of_Children → Discrete feature
- Number_of_Purchases → Discrete feature
- These values are countable.
- Usually, discrete features are whole numbers.
- You cannot normally have 2.5 children or 5.7 purchases.

# 9. Continuous Features

## Definition
Continuous features are numerical features that can have any value within a range, including decimal values.

## Business Example
A manufacturing company may track:
- Product Weight
- Product Length
- Machine Temperature
- Production Time

These can have decimal values, so they are continuous features.

## AI/ML Use
Continuous features are commonly used as numerical inputs in ML models for prediction and pattern detection.

```python

## Python Example

import pandas as pd

data = {
    "Height": [165.5, 170.2, 175.8, 180.4],
    "Weight": [60.5, 68.7, 75.2, 82.6],
    "Temperature": [36.5, 37.1, 36.8, 37.3]
}

df = pd.DataFrame(data)

print(df)

```


## Explanation
- Height → Continuous feature
- Weight → Continuous feature
- Temperature → Continuous feature
- Continuous features can contain decimal values.
- They can take many possible values within a range.

# 10. Identifier Columns

## Definition
An identifier column uniquely identifies each record in a dataset.

## Business Example
A bank dataset may contain:
- Customer_ID → Identifies the customer
- Age → Customer age
- Income → Customer income
- Loan_Approved → Target

Customer_ID is only used to identify the customer, not to predict loan approval.

## AI/ML Use
Identifier columns are usually excluded from model training because they can add noise and do not represent meaningful patterns.

Example:

X = df[["Age", "Salary"]]
y = df["Purchased"]

```python

## Python Example

import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104],
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```


## Explanation

- Customer_ID → Identifier column
- It is used to uniquely identify each customer.
- It usually does not contain useful information for ML prediction.
- Identifier columns are often removed before training a model.

# 11. Target Variable

## Definition
A target variable is the output that a Machine Learning model is trained to predict.

## Business Example
A bank wants to predict whether a customer will repay a loan.

Features:
- Age
- Salary
- Credit Score
- Loan Amount

Target:
- Loan Repaid → Yes/No

## AI/ML Use
The target variable is the value that the ML model learns to predict.

Classification → Target can be Yes/No, Cat/Dog, etc.
Regression → Target can be Salary, Price, Temperature, etc.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Experience": [1, 5, 8, 12],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

# Features
X = df[["Age", "Salary", "Experience"]]

# Target variable
y = df["Purchased"]

print("Target Variable:")
print(y)
```

# 12. Metadata

## Definition
Metadata is information that describes the dataset and its columns.


## Business Example
Before using a customer dataset, a data scientist checks:
- Number of customers
- Column names
- Data types
- Missing values
- Unique values

This helps understand whether the dataset is ready for analysis and ML.

## AI/ML Use
Metadata helps us understand, clean, validate, and prepare a dataset before training a Machine Learning model.
    

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Department": ["IT", "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

# Display dataset metadata
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
```



## Explanation
Metadata tells us about the structure of the dataset.

- Shape → Number of rows and columns
- Columns → Names of the features
- Data Types → Type of data in each column

# 13. Dataset Dimensions

## Definition
Dataset dimensions describe the number of rows and columns in a dataset.

- Rows → Number of records
- Columns → Number of features/variables

## Business Example
An e-commerce dataset has:

- 100,000 customers → Rows
- 20 customer attributes → Columns

Dataset dimensions:

`(100000, 20)`

## AI/ML Use
Dataset dimensions help us understand the size of the data before preprocessing and model training.

```python

## Python Example

import pandas as pd

data = {
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print("Dataset Dimensions:", df.shape)
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

## Explanation
`df.shape` returns:

`(rows, columns)`

Here:

- 4 → Rows
- 4 → Columns

So, the dataset contains **4 records and 4 columns**.
