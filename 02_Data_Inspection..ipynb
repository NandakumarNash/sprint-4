# 1. Number of Rows

## Definition

Number of rows tells us how many records or observations are present in a dataset.

## Business Example

An e-commerce company has customer data.

If the dataset contains 10,000 rows, it means the dataset contains 10,000 customer records.

## AI/ML Use

The number of rows tells us how many observations are available for Machine Learning.

More relevant training data can help a model learn patterns more effectively.

````python

## Python Example

import pandas as pd

# Create sample dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
})

# Find number of rows

print("Number of Rows:")
print(len(df))
````

## Explanation

The dataset contains 4 rows.

Each row represents one customer record.

- Row 1 → Customer 101

- Row 3 → Customer 103
- Row 4 → Customer 104

# 6. head(), tail(), sample()

## Definition

`head()` displays the first few rows of a DataFrame.

`tail()` displays the last few rows of a DataFrame.

`sample()` randomly selects rows from a DataFrame.

These methods are mainly used for quickly inspecting the dataset.


````python

## Python Example

import pandas as pd

# Create sample dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106],
    "Age": [25, 30, 35, 40, 28, 32],
    "Salary": [30000, 50000, 70000, 90000, 45000, 60000],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"]
})

# First 5 rows

print("First 5 Rows:")
print(df.head())

# Last 5 rows

print("\nLast 5 Rows:")
print(df.tail())

# Random 2 rows

print("\nRandom 2 Rows:")
print(df.sample(2))
````

## Business Example

A company receives a customer dataset containing thousands of records.

* head() → Check the first customer records
* tail() → Check the latest/last records
* sample() → Randomly inspect customer records

This helps quickly check whether the dataset looks correct.

## AI/ML Use

These methods are useful during the initial dataset inspection stage.

* head() → Understand the structure and values
* tail() → Check the ending records
* sample() → Randomly inspect data and detect unexpected values

They help us understand the dataset before cleaning and Machine Learning preprocessing.

- `75%` → Third quartile
- `max` → Maximum value

- `max = 90000` → Highest salary is ₹90,000

The `25%`, `50%`, and `75%` values help us understand how the data is distributed.

# 13. Identify Column Types and Roles

## Definition

After inspecting the dataset, we identify the role of each column.

We should identify:

- Numerical columns
- Categorical columns
- Date/Time columns
- Identifier columns
- Potential target columns


## Explanation

The dataset contains different types of columns.

- `Age` → Numerical feature
- `Salary` → Numerical feature
- `Department` → Categorical feature
- `Signup_Date` → Date/Time feature
- `Customer_ID` → Identifier
- `Purchased` → Potential target

Note that `Customer_ID` is technically a numerical column because it contains integers, but its actual role is an identifier.

Therefore, data type and ML role are not always the same.

## Business Example

An e-commerce company wants to predict whether a customer will purchase a product.

The dataset may contain:

- Customer_ID → Identifies the customer
- Age → Customer age
- Salary → Customer income
- Department → Customer department
- Signup_Date → Customer registration date
- Purchased → Whether the customer purchased the product

The company can use the customer information to predict future purchases.

## AI/ML Use

Identifying column roles is an important step before Machine Learning preprocessing.

It helps us decide:

- Which columns can be used as features
- Which columns should be encoded
- Which columns contain dates
- Which columns should be removed as identifiers
- Which column can be used as the target

This prevents incorrect columns from being passed to the Machine Learning model.

````python

## Python Example

import pandas as pd

# Create sample dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Age": [25, 30, 35, 40],
    "Salary": [30000, 50000, 70000, 90000],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Signup_Date": [
        "2025-01-10",
        "2025-02-15",
        "2025-03-20",
        "2025-04-10"
    ],
    "Purchased": ["No", "Yes", "Yes", "Yes"]
})

# Convert date column to datetime

df["Signup_Date"] = pd.to_datetime(df["Signup_Date"])

# Identify numerical columns

numerical_columns = df.select_dtypes(include="number").columns.tolist()

# Identify categorical columns

categorical_columns = df.select_dtypes(include="object").columns.tolist()

# Identify date/time columns

datetime_columns = df.select_dtypes(include="datetime").columns.tolist()

# Identify identifier columns

identifier_columns = ["Customer_ID"]

# Identify potential target column

target_column = "Purchased"

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)

print("\nDate/Time Columns:")
print(datetime_columns)

print("\nIdentifier Columns:")
print(identifier_columns)

print("\nPotential Target Column:")
print(target_column)
````
