# 1. Number of Missing Values

## Definition

The number of missing values tells us how many values are missing in each column of a dataset.

In Pandas, `isnull()` or `isna()` is used to identify missing values, and `sum()` is used to count them.


## Explanation

`isnull()` checks every cell in the DataFrame.

- `True` → value is missing
- `False` → value is present

`sum()` counts the `True` values for each column.

Therefore:

- `Customer_ID` → 0 missing values
- `Age` → 2 missing values
- `Salary` → 2 missing values
- `Department` → 1 missing value

## Business Example

A company can use this check to understand whether customer information is complete.

For example:

- 2 customers have missing `Age`
- 2 customers have missing `Salary`
- 1 customer has a missing `Department`

This helps the company decide whether the missing information needs to be filled, removed, or investigated.

## AI/ML Use

Identifying missing values is an important step in data preprocessing.

Missing values can:

- Cause errors during model training
- Reduce model performance
- Affect statistical calculations
- Introduce bias into the model

Therefore, we should identify the number of missing values before performing feature engineering and model training.

````python

## Python Example

import pandas as pd

# Create sample dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Age": [25, None, 35, 40, None],
    "Salary": [30000, 40000, None, 50000, None],
    "Department": ["IT", "HR", "IT", None, "Finance"]
})

# Count missing values in each column

print("Number of Missing Values:")
print(df.isnull().sum())
````

# 2. Percentage of Missing Values

## Definition

The percentage of missing values tells us what percentage of values are missing in each column of a dataset.

It helps us understand how much missing data exists and decide how to handle it.


## Explanation

`df.isnull()` identifies missing values.

`mean()` calculates the proportion of missing values in each column.

Multiplying by `100` converts the proportion into a percentage.

For example:

- `Customer_ID` → 0% missing
- `Age` → 40% missing
- `Salary` → 40% missing
- `Department` → 20% missing

Here, `Age` has 2 missing values out of 5 records.

So:

`(2 / 5) × 100 = 40%`

## Business Example

An e-commerce company can check customer data quality using the percentage of missing values.

For example:

- `Age` → 40% missing
- `Salary` → 40% missing
- `Department` → 20% missing

A high percentage of missing values may indicate that data collection needs improvement or that the column requires special handling.

## AI/ML Use

Missing-value percentage helps decide how to preprocess features before model training.

For example:

- Low missing percentage → fill missing values
- Moderate missing percentage → investigate and apply an appropriate imputation strategy
- Very high missing percentage → consider removing the feature if it is not useful

Therefore, percentage of missing values is an important data-quality check in an ML pipeline.

````python

## Python Example

import pandas as pd

# Create sample dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Age": [25, None, 35, 40, None],
    "Salary": [30000, 40000, None, 50000, None],
    "Department": ["IT", "HR", "IT", None, "Finance"]
})

# Calculate percentage of missing values

missing_percentage = df.isnull().mean() * 100

print("Percentage of Missing Values:")
print(missing_percentage)
````

# 5. Identify Duplicates

## Definition

Duplicate records are rows that appear more than once in a dataset.

Identifying duplicates is important because duplicate records can make counts inaccurate and may affect data analysis and machine learning models.

In Pandas, `duplicated()` is used to identify duplicate rows.

## Explanation

`duplicated()` checks whether each row has already appeared earlier in the DataFrame.

It returns:

- `False` → row is not a duplicate
- `True` → row is a duplicate

Here, row `3` is marked as `True` because it contains the same patient information as row `1`.

To display only the duplicate rows, we can use:

df[df.duplicated()]

Output:

   Patient_ID Patient_Name  Age Diagnosis
3         102        Priya   32     Fever

By default, Pandas keeps the first occurrence as the original record and marks the later occurrence as a duplicate.

## Business Example

A hospital may accidentally register the same patient twice during data entry.

For example:

Patient_ID = 102  
Patient_Name = Priya  
Age = 32  
Diagnosis = Fever

If this record appears twice, the hospital may incorrectly report:

- 2 patients instead of 1
- 2 consultations instead of 1
- Incorrect treatment statistics
- Incorrect billing records

Therefore, duplicate records should be identified before analysis.

## AI/ML Use

Duplicate records can negatively affect machine learning because the same information may appear multiple times in the training dataset.

This can:

- Give excessive importance to duplicated observations
- Bias model training
- Produce misleading statistics
- Affect model evaluation
- Cause data leakage in some situations

Therefore, identifying duplicate records is an important data-quality step before model training.

````python

## Python Example

import pandas as pd

# Create hospital patient dataset

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 102, 104, 105],
    "Patient_Name": ["Arun", "Priya", "Ravi", "Priya", "Kumar", "Divya"],
    "Age": [45, 32, 50, 32, 60, 28],
    "Diagnosis": ["Diabetes", "Fever", "Asthma", "Fever", "Heart Disease", "Fever"]
})

# Identify duplicate rows

duplicates = df.duplicated()

print("Duplicate Records:")
print(duplicates)
````

# 7. Identify Completely Duplicated Rows

## Definition

Completely duplicated rows are rows where all column values are exactly the same as another row.

This is different from identifying duplicate values in only one column.

In Pandas, `duplicated()` is used to identify completely duplicated rows.

## Explanation

`df.duplicated()` compares the complete row with previously seen rows.

It returns:

- `False` → row is unique
- `True` → complete row is duplicated

Here, row `3` is completely duplicated because all its values are the same as row `1`:

Patient_ID = 102  
Patient_Name = Priya  
Age = 32  
Diagnosis = Fever

The complete row is repeated, so Pandas identifies it as a duplicate.

We use:

df[df.duplicated()]

to display only the completely duplicated rows.

## Business Example

In a hospital database, a patient's registration record may accidentally be entered twice.

For example:

Patient_ID = 102  
Patient_Name = Priya  
Age = 32  
Diagnosis = Fever

If the complete record is entered again, the hospital may count the same patient twice.

This can cause incorrect:

- Patient counts
- Diagnosis statistics
- Billing calculations
- Hospital reports

Identifying completely duplicated rows helps remove such exact duplicate records.

## AI/ML Use

Completely duplicated rows can affect the quality of a machine learning dataset.

They can:

- Increase the frequency of certain observations
- Bias model training
- Distort feature distributions
- Affect statistical analysis
- Increase the risk of data leakage when duplicates appear across training and testing data

Therefore, completely duplicated rows should be identified and reviewed before model training.

````python

## Python Example

import pandas as pd

# Create hospital patient dataset

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 102, 104, 105],
    "Patient_Name": ["Arun", "Priya", "Ravi", "Priya", "Kumar", "Divya"],
    "Age": [45, 32, 50, 32, 60, 28],
    "Diagnosis": ["Diabetes", "Fever", "Asthma", "Fever", "Heart Disease", "Fever"]
})

# Identify completely duplicated rows

duplicate_rows = df[df.duplicated()]

print("Completely Duplicated Rows:")
print(duplicate_rows)
````


# 8. Identify Potential Duplicate Records

## Definition

Potential duplicate records are records that may represent the same entity but are not completely identical.

Unlike completely duplicated rows, some values may be different because of:

- Spelling differences
- Different capitalization
- Formatting differences
- Typing mistakes
- Missing information

Potential duplicates require further investigation before deciding whether they are actually duplicates.

## Python Example

import pandas as pd

df = pd.DataFrame({

print(potential_duplicates)

`duplicated()` can check specific columns instead of the complete row.

```python

Identifying potential duplicates helps an ML Engineer:

# 9. Negative Values Where Inappropriate

## Definition

Negative values are values less than zero.

Negative values are valid for some features, such as temperature or profit and loss, but they are invalid for certain business values such as:

- Age
- Quantity
- Salary
- Number of patients
- Product price
- Blood pressure

We should identify negative values when they are not logically possible for a particular column.


## Explanation

We use a condition such as:

`df["Age"] < 0`

to identify values less than zero.

The condition returns:

- `True` → the value is negative
- `False` → the value is zero or positive

In this example:

- `Age = -5` and `Age = -2` are invalid because a patient's age cannot normally be negative.
- `Bill_Amount = -5000` is unexpected if the column represents the amount charged to a patient.
- `Heart_Rate = -10` is invalid because heart rate cannot normally be negative.

However, negative values are not always invalid.

For example, a column such as `Profit_Loss` may legitimately contain negative values because a negative amount can represent a business loss.

## Business Example

A hospital stores patient billing information.

Suppose one patient's bill is recorded as:

`Bill_Amount = -5000`

This could indicate:

- Data-entry error
- Incorrect transaction
- Refund represented incorrectly
- System integration problem

Similarly, an age of `-5` is clearly invalid and should be investigated before using the data.

## AI/ML Use

Invalid negative values can affect machine learning models because the model may learn from incorrect information.

For example:

- Negative age can distort age-related patterns
- Negative bill amounts can affect financial predictions
- Negative medical measurements can create unrealistic relationships

Therefore, an ML Engineer should apply domain-based validation rules to identify values that are logically impossible before training the model.

````python
## Python Example

import pandas as pd

# Create hospital patient dataset

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, -5, 32, 60, -2],
    "Bill_Amount": [25000, 18000, -5000, 75000, 32000],
    "Heart_Rate": [80, 75, 90, -10, 85]
})

# Check negative values in Age

negative_age = df[df["Age"] < 0]

print("Negative Age Values:")
print(negative_age)

# Check negative values in Bill_Amount

negative_bill = df[df["Bill_Amount"] < 0]

print("\nNegative Bill Amounts:")
print(negative_bill)

# Check negative values in Heart_Rate

negative_heart_rate = df[df["Heart_Rate"] < 0]

print("\nNegative Heart Rate Values:")
print(negative_heart_rate)
````

# 10. Impossible Values

## Definition

Impossible values are values that cannot logically exist based on the meaning or business rules of a column.

These values may be within the expected data type but still be unrealistic or logically invalid.

For example:

- Age = 250
- Exam Mark = 150 when the maximum is 100
- Employee Experience = 60 years
- Temperature = 500°C when the sensor range is limited to 100°C

We identify impossible values using domain-specific validation rules.

## Explanation

We define logical rules based on the meaning of each column.

For `Age`:

`Age < 0` or `Age > 100`

means the age is considered impossible for this dataset.

For `Marks`:

`Marks < 0` or `Marks > 100`

means the mark is outside the valid range.

For `Attendance_Percentage`:

`Attendance_Percentage < 0` or `Attendance_Percentage > 100`

means the value is impossible because a percentage cannot normally exceed 100.

In this dataset, Student 103 has:

- Age = 150
- Marks = 105
- Attendance = 110

All three values violate the defined business rules.

## Business Example

A college can use these validation rules before generating student performance reports.

For example, if a student has:

`Marks = 105`

the value may have been entered incorrectly.

Possible reasons could be:

- Data-entry mistake
- Incorrect data import
- Calculation error
- System integration problem

The college should investigate the original source instead of blindly using the value.

## AI/ML Use

Impossible values can negatively affect machine learning models because the model may learn patterns from unrealistic data.

For example, an age of `150` can:

- Distort feature distributions
- Affect statistical measures
- Create unrealistic relationships
- Reduce model performance

Therefore, ML Engineers use domain rules and validation checks to detect impossible values before feature engineering and model training.

````python

## Python Example

import pandas as pd

# Create student dataset

df = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105],
    "Age": [20, 22, 150, 21, 19],
    "Marks": [85, 92, 105, 78, 88],
    "Attendance_Percentage": [90, 85, 110, 95, 88]
})

# Identify impossible Age values

invalid_age = df[(df["Age"] < 0) | (df["Age"] > 100)]

print("Impossible Age Values:")
print(invalid_age)

# Identify impossible Marks

invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print("\nImpossible Marks:")
print(invalid_marks)

# Identify impossible Attendance values

invalid_attendance = df[
    (df["Attendance_Percentage"] < 0) |
    (df["Attendance_Percentage"] > 100)
]

print("\nImpossible Attendance Values:")
print(invalid_attendance)
````

# 11. Incorrect Categories

## Definition

Incorrect categories are values that do not belong to the valid set of categories defined for a categorical column.

For example, if a `Department` column should contain only `IT`, `HR`, `Finance`, and `Sales`, then values such as `it`, `Human Resources`, or `Fin` may be incorrect or inconsistent categories.

Incorrect categories can occur because of spelling mistakes, different naming conventions, abbreviations, extra spaces, or incorrect data entry.
## Explanation

`isin()` checks whether each value belongs to the list of valid categories.

`df["Department"].isin(valid_departments)` returns:

- `True` → valid category
- `False` → invalid category

The `~` operator reverses the result.

Therefore:

`~df["Department"].isin(valid_departments)`

returns only the rows where the department is not present in the valid category list.

In this example:

- `IT` → valid
- `HR` → valid
- `Finance` → valid
- `it` → incorrect because of lowercase letters
- `Human Resources` → incorrect because the expected category is `HR`
- `Sales` → valid

## Business Example

In an HR system, the company may define departments as:

`IT, HR, Finance, Sales`

But employees may enter:

- `it`
- `IT `
- `Human Resources`
- `Fin`
- `Finance`

Some of these may represent the same department but are stored differently.

For example, `HR` and `Human Resources` may refer to the same department.

If these values are not standardized, the company may produce incorrect reports such as treating `HR` and `Human Resources` as two separate departments.

## AI/ML Use

Incorrect categories can create problems during categorical feature encoding.

For example, if `HR` and `Human Resources` represent the same department, a machine learning model may treat them as two different categories.

This can:

- Increase the number of features
- Create unnecessary sparse data
- Distort feature distributions
- Reduce model quality
- Produce inconsistent predictions

Therefore, ML Engineers should validate and standardize categorical values before encoding and model training.

````python


## Python Example

import pandas as pd

# Create employee dataset

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Employee_Name": ["Arun", "Priya", "Ravi", "Kumar", "Divya", "Manoj"],
    "Department": ["IT", "HR", "Finance", "it", "Human Resources", "Sales"]
})

# Define valid categories

valid_departments = ["IT", "HR", "Finance", "Sales"]

# Identify incorrect categories

invalid_categories = df[
    ~df["Department"].isin(valid_departments)
]

print("Incorrect Categories:")
print(invalid_categories)
````

# 12. Invalid Dates

## Definition

Invalid dates are date values that are incorrect, impossible, or do not follow the expected date format.

Examples include:

- `31/02/2026` → February does not have 31 days
- `2026-13-10` → Month 13 does not exist
- `2026-00-15` → Month 0 is invalid
- Text such as `"unknown"` or `"not available"` in a date column
- Dates written in inconsistent formats

Invalid dates can cause incorrect analysis, sorting, filtering, and time-based calculations.


## Explanation

`pd.to_datetime()` converts a column into Pandas datetime format.

`errors="coerce"` tells Pandas:

- Valid date → convert to datetime
- Invalid date → convert to `NaT`

`NaT` means **Not a Time** and represents a missing or invalid datetime value.

In this example:

- `2024-05-10` → valid
- `2024-02-30` → invalid because February does not have 30 days
- `2024-07-15` → valid
- `2024-13-10` → invalid because month 13 does not exist
- `unknown` → invalid because it is not a valid date

We then use:

`df["Joining_Date"].isna()`

to identify the rows where the date conversion failed.

## Business Example

An HR system stores employee joining dates.

Suppose an employee's joining date is recorded as:

`2024-02-30`

This is an impossible date.

If the company uses this data for calculating employee experience, the result may be incorrect.

Similarly, an invalid joining date can affect:

- Employee tenure
- Payroll calculations
- Leave eligibility
- Experience reports
- Retirement calculations

Therefore, dates should be validated before using them for business calculations.

## AI/ML Use

Invalid dates can affect machine learning features created from date columns.

For example, an ML Engineer may create:

- Employee experience in years
- Customer account age
- Days since last purchase
- Time between transactions

If the original dates are invalid, these calculated features can also become incorrect.

Therefore, date validation is important before creating time-based features for machine learning.

````python

## Python Example

import pandas as pd

# Create employee joining date dataset

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Employee_Name": ["Arun", "Priya", "Ravi", "Kumar", "Divya"],
    "Joining_Date": [
        "2024-05-10",
        "2024-02-30",
        "2024-07-15",
        "2024-13-10",
        "unknown"
    ]
})

# Convert values to datetime
# Invalid values are converted to NaT

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce" 
)

# Identify invalid dates

invalid_dates = df[df["Joining_Date"].isna()]

print("Invalid Dates:")
print(invalid_dates)
````

# 13. Incorrect Data Types

## Definition

Incorrect data types occur when a column contains values stored in a data type that does not match the actual meaning of the data.

For example:

- Age stored as text instead of integer
- Salary stored as text instead of numeric
- Transaction date stored as text instead of datetime
- Quantity stored as text instead of integer

Incorrect data types can cause errors during calculations, filtering, sorting, and machine learning preprocessing.


## Explanation

`df.dtypes` shows the data type of each column.

Initially:

- `Quantity` is stored as `object`
- `Price` is stored as `object`

Although the values look like numbers, they are actually stored as text.

We use:

`pd.to_numeric()`

to convert the text values into numeric values.

After conversion:

- `Quantity` → `int64`
- `Price` → `int64`

Now mathematical operations can be performed correctly.

For example:

`df["Quantity"] * df["Price"]`

can now calculate the total price for each order.

requires both columns to contain numeric values.

# 15. Empty Strings

## Definition

Empty strings are values that contain no characters, represented as `""` or sometimes as strings containing only spaces such as `"   "`.

They are different from `NaN` or `None` because they are technically stored as text values.

Empty strings can occur when users leave a field blank during data entry.

## Explanation

`df == ""` checks for values that are exactly empty strings.

It returns:

- `True` → value is an empty string
- `False` → value is not an empty string

However, a value such as `"   "` contains spaces, so it is not equal to `""`.

To identify both empty strings and space-only values, we use:

`column.astype(str).str.strip().eq("")`

Here:

- `astype(str)` converts values to strings
- `str.strip()` removes leading and trailing spaces
- `eq("")` checks whether the result is empty

Therefore, both `""` and `"   "` can be identified as empty values.

## Business Example

An online shopping website collects customer information.

A customer may leave the `Email` or `City` field blank during registration.

For example:

`Email = ""`

or:

`City = "   "`

These values may look like data is present because they are stored as strings, but they actually contain no useful information.

This can affect:

- Customer communication
- Marketing analysis
- Customer segmentation
- Location-based reporting

## AI/ML Use

Empty strings can create problems when preparing categorical features for machine learning.

For example, if a customer's `City` is stored as `""`, an encoding method may treat the empty string as a separate category.

This can:

- Create meaningless categories
- Increase the number of encoded features
- Affect feature distributions
- Reduce data quality

Therefore, empty strings should be identified and converted to an appropriate missing-value representation, such as `NaN`, before further preprocessing.

````python

## Python Example

import pandas as pd

# Create customer dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Customer_Name": ["Arun", "", "Priya", "   ", "Kumar"],
    "City": ["Chennai", "Bangalore", "", "Mumbai", "Delhi"],
    "Email": ["arun@gmail.com", "", "priya@gmail.com", "   ", "kumar@gmail.com"]
})

# Identify empty strings

empty_string_counts = (df == "").sum()

print("Empty String Count:")
print(empty_string_counts)

# Identify cells containing only spaces

space_only_counts = df.apply(
    lambda column: column.astype(str).str.strip().eq("").sum()
)

print("\nEmpty or Space-only Values:")
print(space_only_counts)
````

# 16. Null-like Values

## Definition

Null-like values are values that represent missing or unavailable data but are not stored as actual `NaN`, `None`, or `NaT` values.

Common null-like values include:

- `"N/A"`
- `"NA"`
- `"None"`
- `"null"`
- `"Unknown"`
- `"-"`
- `"Not Available"`
- Empty strings such as `""`

These values can cause problems because Pandas may treat them as normal text instead of missing values.

## Explanation

A value such as `"N/A"` is stored as text.

For example:

`"N/A"`

is a string, not an actual missing value.

Therefore, simply using:

`df.isnull()`

may not identify it as missing.

We first define the possible null-like values:

`null_like_values = ["N/A", "NA", "None", "null", "Unknown", "-", "Not Available", ""]`

Then:

`df.replace(null_like_values, pd.NA)`

replaces those values with Pandas' actual missing-value representation, `pd.NA`.

After that, `isnull().sum()` can correctly count them as missing values.

## Business Example

A customer-service company may receive customer information from different systems.

One system may use:

`"N/A"`

another may use:

`"Unknown"`

and another may use:

`"None"`

to indicate that the information is unavailable.

For example:

`City = "Unknown"`

does not mean the customer actually lives in a city called "Unknown".

It means the customer's city was not available.

If these values are not standardized, the company may incorrectly count `"Unknown"` as a real category.

## AI/ML Use

Null-like values are important during data preprocessing because machine learning algorithms generally should not treat values such as `"Unknown"` or `"N/A"` as normal categories unless that is intentionally designed.

Unprocessed null-like values can:

- Create incorrect categorical features
- Increase the number of categories
- Affect missing-value analysis
- Cause errors during numerical conversion
- Reduce model quality

Therefore, ML Engineers should identify and standardize null-like values before performing missing-value handling, encoding, and model training.

````python

## Python Example

import pandas as pd

# Create customer dataset

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106],
    "Age": [25, "N/A", 35, "Unknown", 40, 30],
    "City": ["Chennai", "NA", "Bangalore", "Mumbai", "Unknown", "Delhi"],
    "Email": ["arun@gmail.com", "None", "priya@gmail.com", "-", "N/A", "kumar@gmail.com"]
})

# Define common null-like values

null_like_values = ["N/A", "NA", "None", "null", "Unknown", "-", "Not Available", ""]

# Replace null-like values with actual missing values

df_cleaned = df.replace(null_like_values, pd.NA)

print("Dataset After Converting Null-like Values:")
print(df_cleaned)

print("\nMissing Values:")
print(df_cleaned.isnull().sum())
````
