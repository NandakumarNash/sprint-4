# Potential Data Problems

The EDA identified several potential data-quality and data-preparation issues that should be reviewed during the next preprocessing sprint.

## 1. Missing Values

Some features may contain missing values.

Potential impact:

- Reduces the number of usable records
- Can affect statistical calculations
- May cause problems for machine learning models

## 2. Duplicate Records

Duplicate customer or transaction records may exist.

Potential impact:

- Customer activity may be counted multiple times
- Sales statistics may become inaccurate
- Model training may be affected

## 3. Invalid Values

Some numerical features may contain unrealistic values.

Examples:

- Negative `Purchase_Amount`
- `Age` outside a reasonable range
- `Discount` greater than 100%
- `Rating` outside the valid rating scale

These values need to be investigated before modeling.

## 4. Incorrect Data Types

Some columns may have inappropriate data types.

Examples:

- Numerical values stored as text
- Categorical values represented inconsistently
- Numeric columns containing non-numeric characters

## 5. Outliers

Potential outliers may exist in features such as:

- `Purchase_Amount`
- `Discount`
- `Age`

An outlier should not automatically be removed because it may represent a genuine observation.

## 6. Skewed Features

Features such as `Purchase_Amount` may have a right-skewed distribution.

This may affect algorithms that are sensitive to feature distributions.

## 7. Categorical Features

Features such as:

- `Gender`
- `City`
- `Product_Category`
- `Payment_Method`

are categorical and cannot be directly used by many machine learning algorithms.

They may require encoding during preprocessing.

## 8. Different Feature Scales

Numerical features may have very different ranges.

For example:

- `Age` may range from tens to around a hundred.
- `Purchase_Amount` may range from hundreds to thousands.
- `Discount` may be represented as percentages.

Some machine learning algorithms may therefore require feature scaling.

## Conclusion

These issues are documented as **potential preprocessing requirements**.

They should be investigated and handled systematically during the next preprocessing sprint.

No complete preprocessing is performed in this EDA notebook.

# Recommendations

Based on the EDA findings, the following steps are recommended for the next **Data Preprocessing Sprint**.

## 1. Missing-Value Treatment

- Identify columns containing missing values.
- Determine the percentage of missing values.
- Choose an appropriate treatment such as removal or imputation.
- Avoid blindly replacing all missing values with the mean.

## 2. Duplicate Treatment

- Identify duplicate records.
- Determine whether they are genuine repeated transactions or accidental duplicates.
- Remove duplicates only when they are confirmed to be invalid.

## 3. Invalid-Value Treatment

- Identify unrealistic values such as negative purchase amounts or invalid ratings.
- Verify whether they are data-entry errors or genuine observations.
- Correct or remove invalid records when appropriate.

## 4. Data-Type Correction

- Convert numerical columns to appropriate numeric data types.
- Convert categorical columns to appropriate categorical or string types.
- Ensure dates, if present, use proper date formats.

## 5. Outlier Treatment

- Investigate identified outliers.
- Determine whether they represent genuine observations.
- Apply suitable treatment only when justified.

Possible techniques include:

- Capping
- Transformation
- Removal
- Robust scaling

## 6. Encoding

Categorical features such as `Gender`, `City`, `Product_Category`, and `Payment_Method` may need encoding.

Possible techniques include:

- Label Encoding
- One-Hot Encoding
- Ordinal Encoding

The appropriate method should depend on the feature and machine learning algorithm.

## 7. Scaling

Numerical features may require scaling when using algorithms that are sensitive to feature magnitude.

Possible techniques include:

- Standardization
- Min-Max Scaling
- Robust Scaling

## 8. Feature Transformation

Skewed numerical features such as `Purchase_Amount` may require transformation.

Possible approaches include:

- Log transformation
- Power transformation

Transformation should only be applied when it improves the suitability of the data for the intended model.

## 9. Feature Engineering

Potential new features can be created from existing information.

Examples:

- Age Group from `Age`
- Discount Amount from purchase and discount information
- Customer Spending Segment
- Average Purchase by Customer
- High-Value Customer indicator

## 10. Final Recommendation

The next sprint should focus on **cleaning, transforming, and preparing the dataset for machine learning** based on the problems identified during EDA.

The preprocessing process should be documented carefully so that the original data remains traceable.
