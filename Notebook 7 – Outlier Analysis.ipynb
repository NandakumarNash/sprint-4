# 1. What is an Outlier?

### Definition

An **outlier** is a data value that is **significantly different from most other values** in a dataset.

### Real-World Example

Suppose hospital patient bills are:

`₹2,000, ₹2,500, ₹3,000, ₹2,800, ₹3,200, ₹2,700, ₹50,000`

Most bills are around ₹2,000–₹3,200, but **₹50,000 is much higher** than the other bills.

Therefore, `₹50,000` may be an **outlier**.

### Important Point

An outlier is **not automatically an error**.

The value `₹50,000` could be:

- A genuine bill for major surgery → **Retain**
- A billing/data-entry mistake → **Correct or remove**
- An unusual but valid treatment → **Retain**
- An extreme value affecting an ML model → **Consider transformation**

### AI/ML Importance

Outliers can affect:

- Mean and standard deviation
- Data distribution
- Statistical analysis
- ML model performance
- Linear Regression
- KNN
- K-Means

### Example Interpretation

Instead of saying:

"₹50,000 is an outlier."

Say:

"₹50,000 is much higher than the other hospital bills. It should be investigated to determine whether it is a genuine high-cost treatment or a data-entry error."

### Key Point

**Outlier = an unusually different observation.**

**Never automatically remove every outlier. First investigate why it exists, then decide whether to retain, correct, remove, or transform it.**

```python
### Python Example

import numpy as np

bills = np.array([2000, 2500, 3000, 2800, 3200, 2700, 50000])

print(bills)

```

```python
Here, `50000` is far away from the other values, so it may be an outlier.
```

# 2. Why Outliers Occur

### Definition

Outliers can occur for different reasons. An unusual value does not always mean that the data is wrong.

### Common Reasons

- Data-entry error
- Measurement error
- Genuine unusual event
- Natural variation
- Rare event
- Fraud or abnormal activity
- Different population or group

### Real-World Examples

#### 1. Data-Entry Error

Patient age is recorded as:

`25, 32, 41, 29, 250`

`250` is most likely a data-entry error.

**Action:** Correct or remove after verification.

#### 2. Measurement Error

A temperature sensor normally records:

`28°C, 29°C, 30°C, 31°C`

but suddenly records:

`85°C`

The sensor may have malfunctioned.

**Action:** Investigate the measurement system.

#### 3. Genuine Unusual Event

Hospital bills are:

`₹2,000, ₹2,500, ₹3,000, ₹2,800, ₹50,000`

The `₹50,000` bill may be genuine because the patient underwent major surgery.

**Action:** Retain the value if it is valid.

#### 4. Rare Event

Most online transactions may be:

`₹500, ₹800, ₹1,200, ₹2,000`

but one customer makes a transaction of:

`₹5,00,000`

This could be a genuine large purchase or possible fraud.

**Action:** Investigate before deciding what to do.

### Important Point

The same outlier detection result can have different meanings.

For example:

`₹50,000 hospital bill`

Could be:

- Data-entry error → **Correct**
- Genuine surgery → **Retain**
- Unknown reason → **Investigate**

### AI/ML Use

Understanding why outliers occur helps us decide whether to:

- Remove the value
- Correct the value
- Transform the value
- Keep the value
- Investigate the observation further

### Key Point

**Outliers can occur because of errors, rare events, natural variation, or genuine unusual observations.**

**Always understand the reason behind an outlier before deciding what to do with it.**

```python

### Python Example

import numpy as np

ages = np.array([25, 32, 41, 29, 250])

print(ages)


```

Here, `250` is very different from the other ages.

However, we should not immediately delete it.

We first need to determine **why the value is unusual**.

# 3. Types of Outliers

### Definition

Outliers can be classified into different types depending on **how and where they occur in the dataset**.

The main types are:

- Point Outlier
- Contextual Outlier
- Collective Outlier

### 1. Point Outlier

A **point outlier** is a single observation that is significantly different from the other observations.

### Example

Hospital bills:

`₹2,000, ₹2,500, ₹3,000, ₹2,800, ₹3,200, ₹50,000`

Here, `₹50,000` is much higher than the other bills.

Therefore, `₹50,000` is a **point outlier**.

### 2. Contextual Outlier

A **contextual outlier** is a value that is unusual only in a particular situation or context.

### Example

Normal temperatures in a city:

`28°C, 29°C, 30°C, 31°C`

A temperature of `40°C` may be unusual for that location and season.

But `40°C` may be normal in another location or during a different season.

Therefore, the value is an outlier **depending on the context**.

### 3. Collective Outlier

A **collective outlier** occurs when a group or pattern of observations is unusual, even though individual values may not look like outliers.

### Example

A customer's transactions are normally:

`₹500, ₹700, ₹900, ₹600`

But suddenly the customer makes:

`₹500, ₹500, ₹500, ₹500, ₹500, ₹500`

The individual transactions may not look unusual, but the repeated pattern may indicate unusual activity.

Therefore, the **group of observations** can be considered a collective outlier.

### Simple Comparison

| Type | Meaning | Example |
|---|---|---|
| Point Outlier | One unusual value | Hospital bill = ₹50,000 |
| Contextual Outlier | Unusual in a specific context | 40°C during a normally cool season |
| Collective Outlier | A group/pattern is unusual | Repeated unusual transactions |

### AI/ML Importance

Understanding the type of outlier helps us choose the appropriate detection technique.

For example:

- Point outliers → IQR, Z-Score, Box Plot
- Contextual outliers → Analyze context such as time, location, or season
- Collective outliers → Analyze patterns or sequences

### Key Point

**Point outlier = one unusual observation.**

**Contextual outlier = unusual depending on the situation.**

**Collective outlier = a group or pattern of observations is unusual.**

# 4. Univariate Outliers

### Definition

A **univariate outlier** is an unusual value detected by looking at **one numerical variable at a time**.

We examine only one feature and identify values that are very different from the rest.

### Real-World Example

Suppose the salaries of employees are:

`₹25,000, ₹28,000, ₹30,000, ₹32,000, ₹29,000, ₹2,00,000`

Most employees earn between ₹25,000 and ₹32,000, but one employee earns ₹2,00,000.

Therefore, `₹2,00,000` may be a **univariate outlier** in the Salary column.

### Why Is It Called Univariate?

**Uni = One**

**Variate = Variable**

So, **Univariate Outlier Analysis means checking one variable at a time.**

For example:

- Salary → check for unusual salaries
- Age → check for unusual ages
- Hospital Bill → check for unusual bills
- Temperature → check for unusual temperatures

### Example

If we check only the `Salary` column:

| Employee | Salary |
|---|---:|
| A | ₹25,000 |
| B | ₹28,000 |
| C | ₹30,000 |
| D | ₹32,000 |
| E | ₹29,000 |
| F | ₹2,00,000 |

Here, we are using only **Salary** to identify the unusual value.

Therefore, this is **univariate outlier detection**.

### Common Detection Methods

Univariate outliers can be detected using:

- IQR Method
- Z-Score Method
- Box Plot

### Important Point

A univariate outlier is detected using **one variable only**.

We do not consider relationships between multiple variables.

### AI/ML Use

Univariate outlier detection is useful during EDA to:

- Identify unusual values
- Find possible data-entry errors
- Understand data distribution
- Detect extreme observations
- Decide whether transformation may be required

### Key Point

**Univariate Outlier = an unusual observation detected by analyzing one variable at a time.**

```python

### Python Example

import numpy as np

salaries = np.array([25000, 28000, 30000, 32000, 29000, 200000])

print(salaries)
```


The value `200000` is much higher than the other salary values.

Therefore, it may be a univariate outlier.

# 5. Multivariate Outliers

### Definition

A **multivariate outlier** is an observation that may not look unusual when variables are examined individually, but becomes unusual when **two or more variables are considered together**.

### Real-World Example

Suppose we have student data:

| Student | Study Hours | Exam Score |
|---|---:|---:|
| A | 2 | 50 |
| B | 3 | 60 |
| C | 4 | 70 |
| D | 5 | 80 |
| E | 10 | 20 |

Student E has:

- Study Hours = `10`
- Exam Score = `20`

`10` study hours alone may not be an outlier.

`20` exam score alone may not be an outlier.

But **10 study hours with only 20 marks** is unusual compared with the general relationship between Study Hours and Exam Score.

Therefore, Student E may be a **multivariate outlier**.

### Univariate vs Multivariate

**Univariate:**

Check one variable.

Example:

`Salary = ₹2,00,000`

may be unusual when looking only at Salary.

**Multivariate:**

Check multiple variables together.

Example:

`Experience = 15 years` and `Salary = ₹25,000`

may be unusual because the combination does not match the general relationship between Experience and Salary.

### Simple Comparison

| Type | Variables Considered | Example |
|---|---|---|
| Univariate Outlier | One variable | Salary = ₹2,00,000 |
| Multivariate Outlier | Two or more variables | 15 years experience + ₹25,000 salary |

### Detection

Multivariate outliers can be explored using:

- Scatter Plot
- Correlation analysis
- Multivariate statistical methods
- Distance-based methods

### Important Point

A value may look normal when viewed alone but become an outlier when its **relationship with other variables** is considered.

### AI/ML Use

Multivariate outlier detection is important because ML models often use **multiple features together**.

It can help identify:

- Unusual customer behavior
- Fraudulent transactions
- Abnormal medical cases
- Unusual combinations of features
- Data-quality problems

### Key Point

**Multivariate Outlier = an observation that is unusual when multiple variables are considered together.**

```python

### Python Example

    import pandas as pd

    data = {
        "Study_Hours": [2, 3, 4, 5, 10],
        "Exam_Score": [50, 60, 70, 80, 20]
    }

    df = pd.DataFrame(data)

    print(df)
```

The combination:

`Study Hours = 10` and `Exam Score = 20`

is unusual compared with the relationship between the two variables.

# 6. IQR Method

### Definition

The **IQR (Interquartile Range) Method** is a statistical technique used to detect outliers by looking at the middle 50% of the data.

It uses:

- **Q1 (First Quartile)** → 25th percentile
- **Q2 (Median)** → 50th percentile
- **Q3 (Third Quartile)** → 75th percentile
- **IQR** → Q3 − Q1

### Formula

`IQR = Q3 - Q1`

Lower Bound:

`Q1 - 1.5 × IQR`

Upper Bound:

`Q3 + 1.5 × IQR`

Any value:

- Below the Lower Bound → **Potential outlier**
- Above the Upper Bound → **Potential outlier**

### Real-World Example

Suppose hospital bills are:

`₹1,000, ₹1,200, ₹1,500, ₹1,600, ₹1,800, ₹2,000, ₹10,000`

The value `₹10,000` is much higher than most of the bills and may be detected as an outlier using the IQR method.


### Interpretation

The IQR method creates a normal range using Q1 and Q3.

Any value outside that range is flagged as a **potential outlier**.

For example, if:

`Upper Bound = ₹3,000`

and:

`Hospital Bill = ₹10,000`

then `₹10,000` is detected as a potential outlier.

### Important Point

The IQR method **detects or flags** outliers.

It does **not** tell us whether the outlier is:

- A data-entry error
- A genuine observation
- An unusual but valid observation

We must investigate the reason separately.

### Why IQR Is Useful

IQR is especially useful when the data is:

- Skewed
- Not normally distributed
- Contains extreme values

It is less affected by extreme values than methods based directly on the mean and standard deviation.

### AI/ML Use

IQR can be used during EDA to:

- Detect extreme values
- Identify possible data-quality problems
- Understand feature distributions
- Decide whether further investigation is required
- Prepare data before ML modeling

### Key Point

**IQR Method = Q3 − Q1, and values outside 1.5 × IQR from Q1 and Q3 are flagged as potential outliers.**

**Remember: Detection does not mean automatic removal.**

```python

### Python Example

import numpy as np

bills = np.array([1000, 1200, 1500, 1600, 1800, 2000, 10000])

Q1 = np.percentile(bills, 25)
Q3 = np.percentile(bills, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = bills[(bills < lower_bound) | (bills > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

# 7. Z-Score Method

### Definition

The **Z-Score Method** is a statistical technique used to identify outliers by measuring **how far a value is from the mean in terms of standard deviations**.

### Formula

`Z = (X - Mean) / Standard Deviation`

Where:

- `X` = Individual value
- `Mean` = Average of the data
- `Standard Deviation` = Measure of how spread out the data is

### Common Rule

A common rule is:

- Z-Score between `-3` and `+3` → Usually not considered an outlier
- Z-Score below `-3` → Potential outlier
- Z-Score above `+3` → Potential outlier

### Real-World Example

Suppose employee salaries are:

`₹30,000, ₹32,000, ₹31,000, ₹35,000, ₹33,000, ₹2,00,000`

The salary `₹2,00,000` is much higher than the other salaries.

The Z-Score method can measure how many standard deviations `₹2,00,000` is away from the average salary.

### Detecting Potential Outliers

    outliers = salaries[np.abs(z_scores) > 3]

    print("Potential Outliers:", outliers)

If a value has:

`|Z-Score| > 3`

it is commonly flagged as a potential outlier.

### Interpretation

Suppose an employee's salary has:

`Z-Score = 4.2`

This means the salary is approximately **4.2 standard deviations above the mean**.

Therefore, it is a potential outlier.

### Important Point

The Z-Score method works best when the data is approximately **normally distributed**.

For highly skewed data, the IQR method may be more appropriate.

### IQR vs Z-Score

| Method | Based On | Best Used For |
|---|---|---|
| IQR | Quartiles | Skewed/non-normal data |
| Z-Score | Mean and Standard Deviation | Approximately normal data |

### AI/ML Use

Z-Score can help with:

- Detecting extreme values
- Feature analysis
- Data preprocessing
- Standardization
- Identifying unusual observations

### Key Point

**Z-Score tells us how many standard deviations a value is away from the mean.**

**A common rule is to flag values with `|Z| > 3` as potential outliers, but they should still be investigated before removing them.**

```python

### Python Example

import numpy as np
from scipy.stats import zscore

salaries = np.array([30000, 32000, 31000, 35000, 33000, 200000])

z_scores = zscore(salaries)

print(z_scores)

```


### Output

The output will contain one Z-Score for each salary.

The value corresponding to `₹2,00,000` will have a much larger positive Z-Score than the other salaries.

# 8. Box Plot

### Definition

A **Box Plot** is a visualization used to understand the distribution of numerical data and identify **potential outliers**.

A box plot shows:

- Minimum
- Q1 (First Quartile)
- Median
- Q3 (Third Quartile)
- Maximum
- Potential outliers

### Real-World Example

Suppose hospital patient bills are:

`₹1,000, ₹1,200, ₹1,500, ₹1,600, ₹1,800, ₹2,000, ₹10,000`

Most bills are between ₹1,000 and ₹2,000, while `₹10,000` is much higher.

A box plot can visually show `₹10,000` as a potential outlier.

### How to Read a Box Plot

The **box** represents the middle 50% of the data.

The line inside the box represents the **median**.

The lines extending from the box are called **whiskers**.

Individual points beyond the whiskers are potential **outliers**.

### Simple Structure

    Lower Whisker ── [ Q1 | Median | Q3 ] ── Upper Whisker
                                      •
                                      •
                                  Outliers

### Example Interpretation

If the box plot shows a point far above the upper whisker, we can say:

"The box plot shows a potential high-value outlier in the hospital bill data."

However, this does **not** mean the value is incorrect.

The high bill could represent:

- Major surgery
- Emergency treatment
- Long hospital stay

Therefore, we must investigate the observation.

### Relationship With IQR

Box plots commonly use the **IQR method** to identify potential outliers.

    IQR = Q3 - Q1

    Lower Bound = Q1 - 1.5 × IQR

    Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are displayed as potential outliers.

### Important Point

A box plot is mainly a **visual detection tool**.

It helps us quickly see:

- Distribution
- Median
- Spread
- Skewness
- Potential outliers

### AI/ML Use

Box plots are useful during EDA to:

- Detect extreme values
- Compare distributions
- Identify possible data-quality problems
- Compare numerical features across groups
- Decide whether further outlier investigation is required

### Key Point

**Box Plot = a visual way to understand the distribution of numerical data and identify potential outliers.**

**A point outside the whiskers is a potential outlier, not automatically a bad value.**

```python

### Python Example

import matplotlib.pyplot as plt

bills = [1000, 1200, 1500, 1600, 1800, 2000, 10000]

plt.boxplot(bills)
plt.ylabel("Hospital Bill")
plt.title("Hospital Bill Distribution")
plt.show()
```

# 9. Scatter Plot

### Definition

A **Scatter Plot** is a visualization used to show the relationship between **two numerical variables**.

It can also help identify **potential multivariate outliers** by showing observations that are far away from the general pattern.

### Real-World Example

Suppose we have student data:

| Study Hours | Exam Score |
|---:|---:|
| 2 | 50 |
| 3 | 55 |
| 4 | 65 |
| 5 | 70 |
| 6 | 78 |
| 7 | 82 |
| 10 | 20 |

Most students follow a pattern where exam score increases as study hours increase.

But the student with:

`Study Hours = 10`

and:

`Exam Score = 20`

is far away from the general pattern.

This may be a **multivariate outlier**.

### How to Read the Plot

Each point represents **one observation**.

For example:

`(5, 70)`

means:

- Study Hours = 5
- Exam Score = 70

If most points follow a pattern and one point is far away from that pattern, it may be a potential outlier.

### Example Interpretation

Instead of saying:

"The scatter plot has one point far away."

Say:

"The scatter plot shows that most students have higher exam scores as study hours increase. The observation with 10 study hours and a score of 20 is far from this pattern and may be a multivariate outlier."

### Important Point

A scatter plot is especially useful when investigating the **relationship between two variables**.

It can help identify:

- Potential multivariate outliers
- Positive relationships
- Negative relationships
- No clear relationships
- Clusters
- Unusual observations

### Example

#### Age vs Salary

If most employees follow a pattern where salary increases with experience or age, but one employee has:

`Age = 50`

and:

`Salary = ₹20,000`

that combination may be unusual.

The individual values may not necessarily be outliers by themselves, but their **combination** may be unusual.

### AI/ML Use

Scatter plots are useful during EDA to:

- Understand relationships between features
- Identify potential multivariate outliers
- Detect patterns
- Detect clusters
- Understand feature-target relationships
- Decide which features may be useful for ML

### Key Point

**Scatter Plot = a graph showing the relationship between two numerical variables.**

**A point far away from the general pattern may be a potential multivariate outlier, but it must be investigated before deciding what to do with it.**

```python

### Python Example

import matplotlib.pyplot as plt

study_hours = [2, 3, 4, 5, 6, 7, 10]
exam_scores = [50, 55, 65, 70, 78, 82, 20]

plt.scatter(study_hours, exam_scores)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
```

# 10. For Detected Outliers, Determine

### Definition

After detecting an outlier, we must **investigate it before deciding what to do**.

An outlier should not be automatically removed.

### What to Determine

For each detected outlier, ask:

- Is it a genuine observation?
- Is it a data-entry error?
- Is it an unusual but valid observation?
- Should it be removed?
- Should it be transformed?
- Should it be retained?

### Real-World Example

Suppose hospital bills are:

`₹2,000, ₹2,500, ₹3,000, ₹2,800, ₹3,200, ₹2,700, ₹50,000`

The `₹50,000` bill is detected as a potential outlier.

Now we investigate it.

### Case 1: Genuine Observation

The patient underwent major surgery and the actual bill was `₹50,000`.

**Decision:** Retain the value.

**Reason:** It is unusual but genuine.

### Case 2: Data-Entry Error

The actual bill was `₹5,000`, but someone entered `₹50,000`.

**Decision:** Correct the value.

**Reason:** It is a data-entry error.

### Case 3: Unusual but Valid Observation

The patient received an expensive emergency treatment, making the bill much higher than normal.

**Decision:** Retain the value.

**Reason:** It is unusual but valid.

### Case 4: Extreme Value Affecting the Model

The value is genuine, but extreme values are strongly affecting a particular ML model.

**Decision:** Consider an appropriate transformation if it improves the model and is suitable for the feature.

### Decision Table

| Question | Example | Possible Decision |
|---|---|---|
| Genuine observation? | Major surgery bill = ₹50,000 | Retain |
| Data-entry error? | ₹5,000 entered as ₹50,000 | Correct |
| Unusual but valid? | Rare emergency treatment | Retain |
| Should it be removed? | Confirmed invalid value | Remove |
| Should it be transformed? | Genuine extreme value affecting model | Consider transformation |
| Should it be retained? | Genuine and meaningful observation | Retain |

### Document the Reasoning

Do not simply write:

"₹50,000 is an outlier, so I removed it."

Instead, document:

"₹50,000 was detected as a potential outlier using the IQR method. After checking the hospital records, it was confirmed as a genuine bill for major surgery. Therefore, the observation was retained."

### Important Point

**Outlier detection only identifies unusual observations. It does not tell us automatically what action to take.**

The correct process is:

**Detect → Investigate → Decide → Document**

### AI/ML Importance

This investigation is important because removing valid outliers can cause **loss of important information** and may negatively affect an ML model.

### Key Point

**Do not automatically remove every outlier.**

**First determine whether it is genuine, an error, or unusual but valid. Then decide whether to remove, transform, or retain it and document the reasoning.**
