# Important Patterns

Important patterns are repeated or noticeable behaviors found in the dataset.

In a hospital dataset, examples include:

- Most patients belong to a particular age group.
- One department has more patients than other departments.
- Certain diseases occur more frequently.
- Patient visits are concentrated in specific departments.

The goal is to move from observation to business meaning.

Observation:
"The majority of patients are between 40 and 60 years old."

Insight:
"The hospital may need to focus more resources and preventive-care programs on middle-aged patients."

Key idea:

Pattern → Observation → Insight → Business Action

# Trends

A trend is a general direction or change in data over time.

In a hospital dataset, trends can include:

- Patient visits increasing over time.
- Patient visits decreasing during certain months.
- A particular department showing continuous growth.
- Certain diseases becoming more common over time.

Example:

Observation:
"Patient visits increased from January to June."

Insight:
"The increasing patient volume may indicate growing demand for hospital services."

Business Action:
"The hospital may need to plan additional staff, beds, and resources for future periods."

Key idea:

Trend → Observation → Insight → Business Action

# Relationships

Relationships describe how two or more variables are connected to each other.

In a hospital dataset, examples include:

- Age and medical cost.
- Age and length of hospital stay.
- Number of patients and department.
- Treatment type and recovery time.

Example:

Observation:
"Older patients tend to have longer hospital stays."

Insight:
"Older patients may require more intensive care and longer recovery periods."

Business Action:
"The hospital can use this relationship for better bed, staff, and resource planning."

Key idea:

Relationship → Observation → Insight → Business Action

# Anomalies

Anomalies are unusual observations that are significantly different from the normal pattern in the dataset.

In a hospital dataset, examples include:

- A patient with an unusually high medical cost.
- An unusually long hospital stay.
- A department with a sudden unexpected increase in patients.
- An unusually young or old patient.

Example:

Observation:
"One patient has a much higher treatment cost than most other patients."

Insight:
"The case may require investigation to determine whether the high cost is due to a complex medical condition, special treatment, or a data-entry issue."

Business Action:
"The hospital can investigate unusual cases to identify billing issues, improve cost management, or understand complex treatments."

Key idea:

Anomaly → Investigation → Insight → Business Action

# Potential Business Problems

EDA can reveal problems that may affect business operations, cost, efficiency, or customer satisfaction.

In a hospital dataset, potential business problems include:

- One department having an unusually high patient load.
- Long hospital stays increasing resource usage.
- High treatment costs for certain patient groups.
- Frequent readmissions.
- Uneven distribution of patients across departments.

Example:

Observation:
"Department A has significantly more patients than other departments."

Insight:
"The department may be experiencing higher demand and could face staff or resource shortages."

Business Action:
"Management can review staffing levels, bed availability, and resource allocation for Department A."

Key idea:

EDA Finding → Business Problem → Impact → Possible Action

# Interesting Segments

Segments are meaningful groups of records that share common characteristics.

In a hospital dataset, useful segments can include:

- Patients grouped by age.
- Patients grouped by department.
- Patients grouped by disease.
- Patients grouped by treatment type.
- Patients grouped by medical cost.
- Patients grouped by length of hospital stay.

Example:

Observation:
"Patients above 60 years have longer average hospital stays than younger patients."

Insight:
"Older patients may require more care and monitoring during treatment."

Business Action:
"The hospital can create specialized care plans and allocate additional resources for older patients."

Key idea:

Segment → Compare Groups → Find Difference → Business Insight

# Potential Opportunities

EDA can reveal opportunities to improve business performance, reduce costs, improve services, or better allocate resources.

In a hospital dataset, potential opportunities include:

- Improving services in departments with high patient demand.
- Reducing unnecessary long hospital stays.
- Providing targeted care for high-risk patient groups.
- Optimizing staff and bed allocation.
- Identifying departments where additional services may be needed.

Example:

Observation:
"Department A has consistently high patient demand."

Insight:
"Department A may have an opportunity to improve capacity and reduce waiting time."

Business Action:
"The hospital can consider adding staff, beds, or appointment slots in Department A."

Key idea:

EDA Finding → Opportunity → Expected Benefit → Business Action

# Observation → Insight Examples

The goal of Business Insights is to move beyond describing a chart and explain what the result means for the business.

| Visualization Result | Observation | Business Insight |
|----------------------|-------------|------------------|
| Age Distribution | Most patients are between 40–60 years old. | Focus preventive-care programs and resources on middle-aged patients. |
| Department Distribution | General Medicine has the highest patient count. | Allocate more staff and resources to meet higher demand. |
| Treatment Cost Distribution | A few patients have extremely high medical costs. | Review high-cost cases for cost control and specialized care planning. |
| Length of Stay | Older patients stay longer on average. | Improve bed and staff planning for elderly care. |
| Disease Frequency | Diabetes is one of the most common conditions. | Expand diabetes awareness and long-term treatment programs. |

Key idea:

Don't stop at **"What does the chart show?"**  
Always ask **"Why does this matter for the business?"**
