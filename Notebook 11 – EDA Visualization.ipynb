## Histogram

### Why is a Histogram selected?

A Histogram is used to understand the **distribution of a numerical variable**.

It helps us identify:

* How values are distributed
* Where most values are concentrated
* The center and spread of the data
* Whether the distribution is symmetric or skewed
* Possible unusual values

### Real-World Example

Suppose a hospital dataset contains a numerical column called `Age`.

A Histogram can show how the patients' ages are distributed.

### Interpretation

* The **X-axis** represents Age.
* The **Y-axis** represents the number of patients.
* Each bar represents an age range.
* Taller bars mean more patients fall within that age range.
* Shorter bars mean fewer patients fall within that age range.

### Insight Obtained

The Histogram helps us understand the **overall distribution of patient ages**.

For example, if most values are between 30 and 60, we can say that the dataset mainly contains patients in that age range.

**Important:** We create the Histogram only when the dataset contains a suitable numerical variable. We should not create graphs just for the sake of visualization.


````python

### Python Implementation

import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)

df["Age"].plot(kind="hist", bins=10)

plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.title("Age Distribution of Patients")
plt.show()
````

## Box Plot

### Why is a Box Plot selected?

A Box Plot is used to understand the **spread and distribution of numerical data**.

It helps us identify:

* Median
* Lower and upper quartiles
* Spread of the data
* Possible outliers

### Real-World Example

Suppose a hospital dataset contains an `Age` column.

A Box Plot can help us understand the typical patient age and identify unusually low or high ages.

### Interpretation

* The line inside the box represents the **median**.
* The box represents the **middle 50% of the data**.
* The whiskers show the typical range of values.
* Points outside the whiskers may represent **outliers**.

### Insight Obtained

The Box Plot helps us understand the **central value, spread, and possible outliers** in patient ages.

For example, if a few points appear far away from the main box, those observations may need further investigation.


````python
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import matplotlib.pyplot as plt

plt.boxplot(df["Age"])

plt.ylabel("Age")
plt.title("Box Plot of Patient Age")
plt.show()

````

## Bar Chart

### Why is a Bar Chart selected?

A Bar Chart is used to compare **different categories**.

It helps us understand which category has the highest or lowest value.

### Real-World Example

Suppose a hospital dataset contains a `Department` column.

We can use a Bar Chart to compare the **number of patients in each department**.

### Interpretation

* **X-axis:** Hospital departments
* **Y-axis:** Number of patients
* Each bar represents one department.
* A taller bar means that department has more patients.
* A shorter bar means that department has fewer patients.

### Insight Obtained

The Bar Chart helps us identify which hospital departments have the **highest and lowest number of patients**.

For example, if `General` has the tallest bar, the General department has the highest number of patients in this dataset.

### Important

A **Bar Chart is mainly used for comparing categories**, while a **Histogram is used for numerical data distributions**.


````python

### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import matplotlib.pyplot as plt

df["Department"].value_counts().plot(kind="bar")

plt.xlabel("Department")
plt.ylabel("Number of Patients")
plt.title("Number of Patients by Department")
plt.show()
````

## Count Plot

### Why is a Count Plot selected?

A Count Plot is used to show the **number of observations in each category**.

It is especially useful for analyzing **categorical variables** and checking whether some categories occur more frequently than others.

### Real-World Example

Using our hospital dataset, we can count how many patients belong to each `Department`.

### Interpretation

* **X-axis:** Hospital departments
* **Y-axis:** Number of patients
* Each bar represents one department.
* A taller bar means more patients belong to that department.
* A shorter bar means fewer patients belong to that department.

### Insight Obtained

The Count Plot helps us quickly identify the **most common and least common categories**.

For example, if `General` has the tallest bar, General has the highest number of patients in our dataset.

### Bar Chart vs Count Plot

**Bar Chart:** We usually calculate the values first and then plot them.

**Count Plot:** Seaborn automatically counts how many observations belong to each category.

For example:

```python
df["Department"].value_counts()
```

calculates the counts, while:

```python
sns.countplot(x="Department", data=df)
```

automatically calculates and displays them.


````python

### Python Implementation

import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x="Department", data=df)

plt.xlabel("Department")
plt.ylabel("Number of Patients")
plt.title("Patient Count by Department")
plt.show()

````

## Scatter Plot

### Why is a Scatter Plot selected?

A Scatter Plot is used to understand the **relationship between two numerical variables**.

It helps us identify:

* Whether two variables are related
* Positive or negative relationships
* Patterns or trends
* Possible outliers

### Real-World Example

In our hospital dataset, we can examine the relationship between:

* `Age`
* `Blood_Pressure`

We can check whether blood pressure tends to increase as patient age increases.

### Interpretation

* **X-axis:** Patient Age
* **Y-axis:** Blood Pressure
* Each point represents one patient.
* If the points move upward from left to right, there may be a **positive relationship**.
* If they move downward, there may be a **negative relationship**.
* Randomly scattered points may indicate a weak or no relationship.

### Insight Obtained

The Scatter Plot helps us visually identify whether **patient age and blood pressure are related**.

In our sample data, blood pressure generally increases as age increases, suggesting a **positive relationship** between the two variables.

### Important

A Scatter Plot is appropriate when **both variables are numerical**.

Example:

`Age ↔ Blood Pressure` ✅

`Department ↔ Age` ❌


````python

### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import matplotlib.pyplot as plt

plt.scatter(df["Age"], df["Blood_Pressure"])

plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.title("Age vs Blood Pressure")
plt.show()

````

## Line Plot

### Why is a Line Plot selected?

A Line Plot is mainly used to show **how a numerical value changes across an ordered variable**.

It is useful for identifying:

* Trends
* Increases and decreases
* Changes over time or sequence
* Overall direction of the data

### Real-World Example

Using our hospital dataset, we can observe how `Treatment_Cost` changes with `Age`.

Here, Age is ordered from younger to older patients, so a line can show the trend in treatment cost.

### Interpretation

* **X-axis:** Patient Age
* **Y-axis:** Treatment Cost
* Each point represents a patient.
* The connecting line shows how treatment cost changes as age increases.
* An upward direction indicates increasing treatment cost.
* A downward direction indicates decreasing treatment cost.

### Insight Obtained

The Line Plot helps us identify the **trend in treatment cost across patient ages**.

In our sample data, treatment cost generally increases as age increases, although the relationship is not necessarily perfectly linear.

### Important

A Line Plot is most meaningful when the X-axis has a **natural order**, such as:

* Date → Sales
* Month → Revenue
* Year → Population

For two unrelated numerical variables, a **Scatter Plot** is usually more appropriate.


````python

### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import matplotlib.pyplot as plt

sorted_df = df.sort_values("Age")

plt.plot(sorted_df["Age"], sorted_df["Treatment_Cost"], marker="o")

plt.xlabel("Age")
plt.ylabel("Treatment Cost")
plt.title("Treatment Cost by Age")
plt.show()

````

## Violin Plot

### Why is a Violin Plot selected?

A Violin Plot is used to understand the **distribution and spread of numerical data**.

It combines ideas from a **Box Plot** and a **distribution plot**.

It helps us understand:

* Median
* Spread
* Data density
* Distribution shape
* Possible unusual values

### Real-World Example

Using our hospital dataset, we can visualize the distribution of `Treatment_Cost`.

This helps us understand where treatment costs are concentrated and how widely they vary.

### Interpretation

* **Y-axis:** Treatment Cost
* The wider part of the violin represents an area where more values are concentrated.
* The narrower part represents fewer values.
* The shape shows the overall distribution of treatment costs.
* The middle of the distribution represents the central tendency.

### Insight Obtained

The Violin Plot helps us understand **where treatment costs are concentrated and how the distribution is shaped**.

For example, if the violin is wider around ₹1,000–₹1,800, many patients have treatment costs in that range.

### Box Plot vs Violin Plot

**Box Plot** → Better for quickly identifying median, quartiles, spread, and outliers.

**Violin Plot** → Shows those distribution characteristics while also showing the **density and shape of the data**.


````python

### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(y=df["Treatment_Cost"])

plt.ylabel("Treatment Cost")
plt.title("Distribution of Treatment Cost")
plt.show()

````

## Pair Plot

### Why is a Pair Plot selected?

A Pair Plot is used to examine the **relationships between multiple numerical variables at the same time**.

It helps us identify:

* Relationships between variables
* Positive or negative trends
* Distributions of individual variables
* Possible clusters
* Possible outliers

### Real-World Example

In our hospital dataset, we can examine the relationships between:

* `Age`
* `Blood_Pressure`
* `Treatment_Cost`

Instead of creating separate scatter plots for every pair, a Pair Plot displays them together.

### Interpretation

The Pair Plot creates a matrix of plots.

* Each **column** represents one variable.
* Each **row** represents one variable.
* The **diagonal plots** show the distribution of each individual variable.
* The plots outside the diagonal show relationships between two variables.

For example:

`Age ↔ Blood_Pressure`

`Age ↔ Treatment_Cost`

`Blood_Pressure ↔ Treatment_Cost`

### Insight Obtained

The Pair Plot provides an overall view of relationships between multiple numerical variables.

For our sample data, we can observe that:

* Age and Blood Pressure show an increasing pattern.
* Age and Treatment Cost generally show an increasing pattern.
* Blood Pressure and Treatment Cost also show a general increasing pattern.

### Important

Pair Plot is useful when we have **multiple numerical variables** and want to explore their relationships together.

It can become difficult to read when the dataset contains a very large number of variables, so we should select relevant columns rather than plotting everything.


````python

### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df[["Age", "Blood_Pressure", "Treatment_Cost"]])

plt.show()

````

## Heatmap

### Why is a Heatmap selected?

A Heatmap is used to visualize **relationships between multiple numerical variables** using colors.

In EDA, it is commonly used to visualize a **correlation matrix**.

It helps us quickly identify:

* Strong relationships
* Weak relationships
* Positive correlation
* Negative correlation
* Variables that may be useful for further analysis

### Real-World Example

Using our hospital dataset, we can check the correlation between:

* `Age`
* `Blood_Pressure`
* `Treatment_Cost`

### Interpretation

The values in the heatmap range from **-1 to +1**.

* **+1** → Strong positive relationship
* **0** → No linear relationship
* **-1** → Strong negative relationship

For example, if the value between `Age` and `Blood_Pressure` is close to `+1`, they have a strong positive relationship.

### Insight Obtained

The Heatmap gives us a quick overview of relationships between numerical variables.

In our sample data, we can observe positive relationships between:

* `Age` and `Blood_Pressure`
* `Age` and `Treatment_Cost`
* `Blood_Pressure` and `Treatment_Cost`

### Important

A Heatmap is **not limited to correlation**. It can visualize other matrix-style data as well.

For EDA, the **correlation heatmap** is especially useful for understanding relationships between numerical features before machine learning.


````python

### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import seaborn as sns
import matplotlib.pyplot as plt

correlation = df[["Age", "Blood_Pressure", "Treatment_Cost"]].corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()

````

## Distribution Plot

### Why is a Distribution Plot selected?

A Distribution Plot is used to understand the **shape and distribution of a numerical variable**.

It helps us identify:

* Where most values are concentrated
* The center of the distribution
* Spread of the data
* Skewness
* Overall distribution shape

### Real-World Example

Using our hospital dataset, we can examine the distribution of `Treatment_Cost`.

This helps us understand how treatment costs are distributed among patients.

### Interpretation

* **X-axis:** Treatment Cost
* **Y-axis:** Number of Patients
* The bars show how frequently treatment costs occur.
* The smooth curve shows the overall shape of the distribution.
* A peak indicates where values are concentrated.
* A long tail may indicate skewness.

### Insight Obtained

The Distribution Plot helps us understand the **overall shape of the treatment-cost distribution**.

For example, if most values are concentrated around ₹1,000–₹2,000 and a few values extend toward ₹3,000, the distribution may be **right-skewed**.

### Important

A Distribution Plot and Histogram are closely related.

* **Histogram:** Shows frequency using bars.
* **Distribution Plot:** Shows the distribution and can add a **KDE curve** to visualize the estimated density.

So, we should use the Distribution Plot when we want to understand not only the frequency but also the **shape of the distribution**.


````python
### Python Implementation
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 72],
    "Department": ["General", "Cardiology", "General", "Neurology", "Cardiology",
                    "General", "Orthopedic", "Cardiology", "General", "Neurology",
                    "Orthopedic", "General", "Cardiology", "Neurology", "Orthopedic"],
    "Blood_Pressure": [110, 115, 118, 120, 122, 125, 128, 130, 132, 135, 138, 140, 145, 150, 170],
    "Treatment_Cost": [500, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 3000]
}

df = pd.DataFrame(data)
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["Treatment_Cost"], kde=True)

plt.xlabel("Treatment Cost")
plt.ylabel("Number of Patients")
plt.title("Distribution of Treatment Cost")
plt.show()

````
