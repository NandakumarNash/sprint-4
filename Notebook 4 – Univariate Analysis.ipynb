## 1. What is Univariate Analysis?

### Definition

Univariate analysis means analyzing **one variable/column at a time** to understand its distribution, central tendency, spread, and patterns.

- Is the distribution symmetric or skewed?
- Are there any unusual values or outliers?

### In AI/ML

# 2. Mean

## Definition

Mean is the **average value** of a numerical variable.

It is calculated by adding all values and dividing the total by the number of values.

### Formula

Mean = Sum of all values / Number of values


## Explanation

The patient ages are:

25, 30, 35, 40, 60

Mean:

(25 + 30 + 35 + 40 + 60) / 5

= 190 / 5

= 38

Therefore, the **average patient age is 38 years**.

## Business Example

A hospital wants to understand the average age of its patients.

If the mean age is 38 years, the hospital can say that the average age of patients in the dataset is **38 years**.

For example:

Age = [25, 30, 35, 40, 60]

Mean = 38

This gives the hospital a simple summary of the Age variable.

## AI/ML Use

Mean is useful during:

- Exploratory Data Analysis (EDA)
- Understanding numerical features
- Data preprocessing
- Missing-value handling
- Feature analysis

For example, missing Age values can sometimes be replaced with the mean:

df["Age"] = df["Age"].fillna(df["Age"].mean())

However, mean is sensitive to **outliers**.

Example:

Age = [25, 30, 35, 40, 200]

The value `200` is an extreme value and will increase the mean significantly.

Therefore, before using mean for missing-value imputation, we should check for outliers and the distribution of the feature.

## Finding

> The mean represents the average value of a numerical variable. It is useful for understanding the center of the data, but it can be strongly affected by extreme values or outliers.

````python

## Python Example

import pandas as pd

# Create hospital patient dataset

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [25, 30, 35, 40, 60]
})

# Calculate mean

mean_age = df["Age"].mean()

print("Mean Age:", mean_age)
````

# 3. Median

## Definition

Median is the **middle value** of a numerical variable after arranging the values in ascending or descending order.

If there is an even number of values, the median is the average of the two middle values.

Median is a **measure of central tendency**.


## Explanation

The Age values are:

25, 30, 35, 40, 60

The values are already arranged in ascending order.

The middle value is 35.

Therefore, the median age is **35 years**.

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

If there are an even number of values:

ages = [25, 30, 35, 40]

The two middle values are 30 and 35.

Median = (30 + 35) / 2 = 32.5

Therefore, the median is 32.5.

## Business Example

A hospital wants to understand the typical age of its patients.

Suppose the ages are:

[25, 30, 35, 40, 200]

The median is 35.

The value 200 is an extreme value, but it does not significantly affect the median.

Therefore, the hospital can use the median to understand the typical patient age when extreme values are present.

## AI/ML Use

Median is useful for:

- Exploratory Data Analysis (EDA)
- Understanding numerical features
- Handling skewed data
- Missing-value imputation
- Data preprocessing

For example, missing Age values can be replaced with the median:

df["Age"] = df["Age"].fillna(df["Age"].median())

Median is often preferred over mean when a numerical feature contains **outliers or is highly skewed**.

Example:

Age = [25, 30, 35, 40, 200]

Mean = 66

Median = 35

The extreme value 200 greatly affects the mean, but has much less effect on the median.

## Finding

The median represents the **middle value of one numerical variable**. It is less affected by outliers than the mean, making it useful for skewed data and datasets containing extreme values.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [25, 30, 35, 40, 60]
})

median_age = df["Age"].median()

print("Median Age:", median_age)
````

# 4. Mode

## Definition

Mode is the **most frequently occurring value** in a variable.

For numerical variables, mode tells us which value appears most often.

Mode can also be used for categorical variables.

## Explanation

The Age values are:

25, 30, 30, 35, 40, 30

Frequency of each value:

25 → 1 time

30 → 3 times

35 → 1 time

40 → 1 time

The value **30** occurs most frequently.

Therefore:

Mode Age = 30

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

A dataset can have:

- One mode → Unimodal
- Two modes → Bimodal
- More than two modes → Multimodal
- No repeated value → No mode

## Business Example

A hospital wants to know the most common age among its patients.

Suppose:

Age = [25, 30, 30, 35, 40, 30]

The mode is 30.

This means **30 is the most frequently occurring patient age** in the dataset.

Another example is a hospital analyzing blood groups:

Blood Group = [A, B, A, O, A, AB]

Mode = A

This tells the hospital that **A is the most common blood group** in the dataset.

## AI/ML Use

Mode is useful for:

- Exploratory Data Analysis (EDA)
- Understanding the most common value
- Categorical feature analysis
- Missing-value imputation
- Data preprocessing

For example, a categorical column such as City may contain missing values.

The missing City values can sometimes be replaced with the most frequently occurring city:

df["City"] = df["City"].fillna(df["City"].mode()[0])

This is called **mode imputation**.

Mode is especially useful for categorical variables because mean and median cannot normally be calculated for categories such as:

Chennai, Bangalore, Mumbai, Delhi

## Finding

The mode represents the **most frequently occurring value** in a variable. It is useful for identifying the most common value and is particularly useful for categorical data and missing-value imputation.

# 7. Range

## Definition

Range is the **difference between the maximum and minimum values** of a numerical variable.

### Formula

Range = Maximum - Minimum

Range gives a quick idea about the **overall spread** of the data.

## Explanation

The Age values are:

25, 30, 35, 40, 60

Minimum = 25

Maximum = 60

Therefore:

Range = Maximum - Minimum

Range = 60 - 25

Range = 35

So, the Age variable has a **range of 35 years**.

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

In EDA, range helps us quickly understand how widely the values are spread between the smallest and largest observations.

## Business Example

A hospital wants to understand the age span of its patients.

If:

Minimum Age = 25

Maximum Age = 60

Then:

Age Range = 35 years

This means the difference between the youngest and oldest patient in the dataset is 35 years.

Suppose instead:

Age = [25, 30, 35, 40, 150]

Range = 150 - 25 = 125

The very large range suggests that we should investigate the value 150.

## AI/ML Use

Range is useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature spread
- Detecting potentially unusual values
- Data validation
- Comparing feature scales
- Data preprocessing

However, range depends only on the **minimum and maximum values**.

Therefore, a single extreme value can make the range very large.

For example:

Age = [25, 30, 35, 40, 200]

Range = 200 - 25 = 175

Here, the value 200 greatly increases the range.

Therefore, in EDA, range should be considered along with other measures such as **standard deviation, quartiles, and IQR**.

## Finding

The range shows the **difference between the smallest and largest values of one numerical variable**. In EDA, it provides a quick understanding of the overall spread, but it can be strongly affected by extreme values.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [25, 30, 35, 40, 60]
})

minimum_age = df["Age"].min()
maximum_age = df["Age"].max()

age_range = maximum_age - minimum_age

print("Minimum Age:", minimum_age)
print("Maximum Age:", maximum_age)
print("Age Range:", age_range)

````

# 8. Variance

## Definition

Variance measures **how far the values of a numerical variable are spread out from the mean**.

- Low variance → values are close to the mean
- High variance → values are more spread out

In EDA, variance helps us understand the **variability** of a numerical feature.

## Explanation

The Age values are:

25, 30, 35, 40, 60

Mean Age = 38

The values are not all the same; they are spread around the mean.

Variance measures this spread by calculating the squared differences between each value and the mean.

In EDA, we mainly use variance to understand whether a feature has **low or high variability**.

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

### Important Note

Variance is expressed in **squared units**.

For example:

Age → years

Variance → years²

Therefore, standard deviation is often easier to interpret because it is in the original unit.

## Business Example

A hospital analyzes the ages of patients in two departments.

Department A:

Age = [35, 36, 35, 34, 36]

The ages are very close together, so the variance is low.

Department B:

Age = [20, 30, 40, 50, 60]

The ages are more spread out, so the variance is higher.

This tells the hospital that Department B has greater variation in patient age.

## AI/ML Use

Variance is useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature variability
- Detecting features with very little variation
- Feature selection
- Data preprocessing

For example, if a feature has almost the same value for every row:

Age = [30, 30, 30, 30, 31]

Its variance is very low.

Such a feature may provide very little useful information to a machine learning model.

### Important EDA Point

A feature with **zero variance** has the same value for every observation.

Example:

Department = ["Cardiology", "Cardiology", "Cardiology", "Cardiology"]

There is no variation in the feature.

Such a feature may be removed during preprocessing because it does not help distinguish between observations.

## Finding

Variance tells us **how much a numerical variable varies around its mean**. In EDA, high variance indicates greater spread, while very low or zero variance indicates little or no variation in the feature.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [25, 30, 35, 40, 60]
})

variance_age = df["Age"].var()

print("Age Variance:", variance_age)
````

# 9. Standard Deviation

## Definition

Standard deviation measures **how much the values of a numerical variable typically differ from the mean**.

- Low standard deviation → values are close to the mean
- High standard deviation → values are widely spread from the mean

In EDA, standard deviation helps us understand the **spread and variability** of a numerical feature.

## Explanation

The Age values are:

25, 30, 35, 40, 60

Mean Age = 38

The values are spread around the mean.

Standard deviation tells us approximately how far the Age values are spread from the mean.

Here, the standard deviation is approximately **13.15 years**.

This means the ages typically vary around the mean age of 38 by roughly 13 years.

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

### Variance vs Standard Deviation

Variance:

- Measures spread
- Expressed in squared units
- Age → years²

Standard deviation:

- Measures spread
- Expressed in the original unit
- Age → years

Therefore, standard deviation is usually easier to interpret.

## Business Example

A hospital compares the ages of patients in two departments.

Department A:

Age = [35, 36, 35, 34, 36]

The ages are close together.

→ Low standard deviation

Department B:

Age = [20, 30, 40, 50, 60]

The ages are widely spread.

→ High standard deviation

This helps the hospital understand how much patient ages vary within each group.

## AI/ML Use

Standard deviation is useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature variability
- Detecting unusual values
- Feature scaling
- Data preprocessing

Standard deviation is also important for **standardization**.

For example, a feature can be standardized using:

z = (x - mean) / standard deviation

In Python, StandardScaler is commonly used:

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df["Age_scaled"] = scaler.fit_transform(df[["Age"]])

This transforms the feature so that it is centered around 0 with a standard deviation close to 1.

## Finding

Standard deviation tells us **how much a numerical variable varies around its mean**. In EDA, a high standard deviation indicates greater spread, while a low standard deviation indicates that the values are concentrated closer to the mean.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [25, 30, 35, 40, 60]
})

std_age = df["Age"].std()

print("Age Standard Deviation:", std_age)
````

# 10. Quartiles

## Definition

Quartiles divide a sorted numerical variable into **four equal parts**.

They help us understand how the data is distributed across different sections.

The main quartiles are:

- **Q1 (25th percentile)** → 25% of the data is at or below this value
- **Q2 (50th percentile)** → 50% of the data is at or below this value; this is the median
- **Q3 (75th percentile)** → 75% of the data is at or below this value

The difference between Q3 and Q1 is called the **Interquartile Range (IQR)**.

Formula:

IQR = Q3 - Q1

## Explanation

The Age values are:

20, 25, 30, 35, 40, 45, 50, 60

Quartiles divide the data into four sections.

Q1 = 26.25

This means approximately **25% of the observations are below or equal to 26.25**.

Q2 = 37.5

This is the median and represents the **50th percentile**.

Q3 = 46.25

This means approximately **75% of the observations are below or equal to 46.25**.

IQR = Q3 - Q1

IQR = 46.25 - 26.25

IQR = 20

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

## Business Example

A hospital wants to understand how patient ages are distributed.

Suppose:

Q1 = 26.25

Q2 = 37.5

Q3 = 46.25

The hospital can understand that the middle 50% of patients have ages between approximately **26.25 and 46.25 years**.

This gives more information than simply looking at the average age.

## AI/ML Use

Quartiles are useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature distribution
- Detecting outliers
- Understanding skewed data
- Creating box plots
- Data preprocessing

IQR is commonly used for **outlier detection**.

The standard IQR method uses:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound may be considered potential outliers.

For example:

Q1 = 26.25
Q3 = 46.25
IQR = 20

Lower Bound = 26.25 - (1.5 × 20) = -3.75

Upper Bound = 46.25 + (1.5 × 20) = 76.25

Therefore, an Age value greater than 76.25 would be flagged as a potential outlier.

## Finding

Quartiles divide one numerical variable into four parts and help us understand its distribution. In EDA, **Q1, Q2, Q3, and IQR** are especially useful for understanding the middle 50% of the data and detecting potential outliers.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Age": [20, 25, 30, 35, 40, 45, 50, 60]
})

Q1 = df["Age"].quantile(0.25)
Q2 = df["Age"].quantile(0.50)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

````

# 11. Percentiles

A percentile tells us the **position of a value relative to the rest of the data**.

For example:

- 25th percentile → 25% of the data is at or below this value
- 50th percentile → 50% of the data is at or below this value
- 75th percentile → 75% of the data is at or below this value
- 90th percentile → 90% of the data is at or below this value
- 95th percentile → 95% of the data is at or below this value

In EDA, percentiles help us understand the **distribution and position of values** in a numerical feature.

## Explanation

The 25th percentile means approximately **25% of patients are at or below 26.25 years**.

The 50th percentile means approximately **50% of patients are at or below 37.5 years**.

The 75th percentile means approximately **75% of patients are at or below 46.25 years**.

The 90th percentile means approximately **90% of patients are at or below 53 years**.

### Quartiles vs Percentiles

Quartiles are specific percentile points:

Q1 = 25th percentile

Q2 = 50th percentile

Q3 = 75th percentile

So, **quartiles are a subset of percentiles**.

## Business Example

A hospital wants to understand the age distribution of its patients.

Suppose the 90th percentile of Age is 53 years.

Percentiles are useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature distribution
- Detecting extreme values
- Outlier analysis
- Data preprocessing
- Setting thresholds

This means approximately **95% of transactions are ₹50,000 or below**.

The company can investigate transactions above ₹50,000 as unusually high transactions.

Percentiles are also useful for identifying extreme values without relying only on the mean.

## Finding

Percentiles tell us **where a value lies within the distribution of one numerical variable**. In EDA, they help us understand the distribution, identify extreme values, and create useful thresholds for analysis or preprocessing.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Age": [20, 25, 30, 35, 40, 45, 50, 60]
})

p25 = df["Age"].quantile(0.25)
p50 = df["Age"].quantile(0.50)
p75 = df["Age"].quantile(0.75)
p90 = df["Age"].quantile(0.90)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
print("90th Percentile:", p90)
````

# 12. Skewness

## Definition

Skewness measures the **asymmetry of a numerical variable's distribution**.

It tells us whether the data is distributed evenly around the center or has a longer tail on one side.

There are three common cases:

- **Skewness ≈ 0** → Approximately symmetric distribution
- **Positive skewness** → Right-skewed; longer tail on the right
- **Negative skewness** → Left-skewed; longer tail on the left

In EDA, skewness helps us understand the **shape of a numerical feature's distribution**.

## Explanation

The Hospital_Bill values are:

1000, 1200, 1500, 1800, 2000, 15000

Most bills are relatively low, but one bill is extremely high.

The high value `15000` creates a **long tail toward the right**.

Therefore, the distribution is **positively skewed (right-skewed)**.

Here, we are analyzing only the **Hospital_Bill** variable, so this is **univariate analysis**.

### Understanding the Three Cases

### 1. Symmetric

Example:

10, 20, 30, 40, 50

The distribution is approximately balanced.

Skewness ≈ 0

### 2. Positive Skew / Right Skew

Example:

1000, 1200, 1500, 1800, 2000, 15000

A few large values create a long right tail.

Skewness > 0

Usually:

Mean > Median

### 3. Negative Skew / Left Skew

Example:

10, 800, 850, 900, 920, 950

A few small values create a long left tail.

Skewness < 0

Usually:

Mean < Median

## Business Example

A hospital analyzes patient treatment bills.

Suppose most patients have bills between ₹1,000 and ₹5,000, but a few patients have bills above ₹1,00,000 because they required expensive surgeries.

The bill distribution will likely be **positively skewed**.

This tells the hospital that a small number of patients have exceptionally high treatment costs.

For such data, the **median** may represent the typical bill better than the mean.

## AI/ML Use

Skewness is useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature distributions
- Detecting asymmetric data
- Identifying extreme values
- Deciding whether transformation may be useful
- Data preprocessing

For example, if a feature such as Income or Hospital_Bill is highly right-skewed, an ML Engineer may consider a transformation such as:

import numpy as np

df["Hospital_Bill_log"] = np.log1p(df["Hospital_Bill"])

A log transformation can reduce strong right skew and make the feature distribution more manageable for some ML algorithms.

However, transformation should be based on the dataset and model requirements, not applied automatically.

## Finding

Skewness tells us **whether one numerical variable is symmetric or has a longer tail on the left or right**. In EDA, it helps identify distribution shape, extreme values, and whether a feature may require further investigation or transformation.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106],
    "Hospital_Bill": [1000, 1200, 1500, 1800, 2000, 15000]
})

skewness = df["Hospital_Bill"].skew()

print("Hospital Bill Skewness:", skewness)
````

# 13. Kurtosis

## Definition

Kurtosis describes the **tail heaviness and presence of extreme values** in a numerical variable's distribution.

In EDA, kurtosis helps us understand whether a feature has **more or fewer extreme values compared with a normal distribution**.

Common interpretation using Pandas' `kurt()`:

- **Kurtosis ≈ 0** → Similar tail behavior to a normal distribution
- **Positive kurtosis** → Heavier tails; more extreme values
- **Negative kurtosis** → Lighter tails; fewer extreme values

## Explanation

The Hospital_Bill values are:

1000, 1200, 1500, 1800, 2000, 15000

Most bills are between ₹1,000 and ₹2,000, but one bill is ₹15,000.

The very high value creates a **heavy tail** and indicates the presence of an extreme observation.

Therefore, the feature has **positive kurtosis**.

Here, we are analyzing only the **Hospital_Bill** variable, so this is **univariate analysis**.

### Important Difference

Skewness and kurtosis tell us different things:

**Skewness:**

> Is the distribution asymmetric? Which side has the longer tail?

**Kurtosis:**

> How heavy are the tails? Are extreme values more prominent?

For example:

A feature can be approximately symmetric but still have high kurtosis because it contains extreme values on both sides.

## Business Example

A hospital analyzes patient treatment bills.

Most patients may have bills between ₹1,000 and ₹5,000, while a few patients may have bills of ₹1,00,000 or more because of major surgeries.

This can produce high positive kurtosis.

The hospital should investigate these extreme bills to determine whether they are:

- Genuine expensive treatments
- Special cases
- Data-entry errors
- Incorrect billing records

## AI/ML Use

Kurtosis is useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature distributions
- Detecting extreme values
- Identifying heavy-tailed features
- Data-quality investigation
- Deciding whether transformations may be useful

For example, if a financial transaction feature has very high kurtosis, it may contain a small number of unusually large transactions.

An ML Engineer should investigate these values before deciding whether to:

- Keep them
- Correct them
- Remove them
- Transform the feature

A transformation such as `log1p()` may sometimes help with highly skewed positive-valued features:

import numpy as np

df["Log_Bill"] = np.log1p(df["Hospital_Bill"])

The transformation should be applied based on the actual distribution and ML problem.

## Finding

Kurtosis helps us understand the **tail heaviness and extreme values of one numerical variable**. In EDA, high positive kurtosis can indicate that the feature contains unusually extreme observations that should be investigated.

# 15. KDE / Density Plot

## Definition

KDE (Kernel Density Estimation) or Density Plot is a **smooth visualization of the distribution of one numerical variable**.

Instead of showing separate bars like a histogram, KDE shows a **smooth curve** that represents where the data is more or less concentrated.

In EDA, KDE helps us understand:

- Shape of the distribution
- Where values are concentrated
- Skewness
- Spread of the data
- Possible multiple peaks

## Explanation

The KDE curve estimates the shape of the Age distribution.

For example:

Age = [20, 25, 28, 30, 32, 35, 40, 60]

Most values are between 20 and 40.

Therefore, the density will be higher around that region.

The value 60 is far from most observations and may create a longer tail toward the right.

This can help us identify a **right-skewed distribution**.

Here, we are analyzing only the **Age** variable, so this is **univariate analysis**.

### Histogram vs KDE

Histogram:

> Uses bars and bins to show frequency.

KDE:

> Uses a smooth curve to show estimated density.

They are often used together in EDA.

## Business Example

A hospital wants to understand the distribution of patient ages.

A KDE plot can show where patient ages are concentrated.

For example, if the curve has its highest concentration around 30–40 years, most patients are around that age range.

If the curve has a long tail toward older ages, the hospital can investigate whether there are relatively fewer but much older patients.

## AI/ML Use

KDE is useful during:

- Exploratory Data Analysis (EDA)
- Understanding feature distribution
- Identifying skewness
- Comparing distribution shapes
- Identifying multiple peaks
- Detecting unusual patterns

For example, if a feature has two clear peaks, it may indicate that the dataset contains **two different groups**.

For a hospital dataset, two peaks in Age might suggest two dominant patient age groups.

KDE can therefore help an ML Engineer understand the structure of a feature before selecting preprocessing or modeling approaches.

## Finding

KDE provides a **smooth estimate of the distribution of one numerical variable**. In EDA, it helps us understand where values are concentrated, identify skewness and multiple peaks, and visually inspect the shape of the feature distribution.

````python

## Python Example

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Age": [20, 25, 28, 30, 32, 35, 40, 60]
})

sns.kdeplot(df["Age"], fill=True)

plt.xlabel("Age")
plt.ylabel("Density")
plt.title("Age Density Distribution")

plt.show()
````

# 16. Box Plot

## Definition

A Box Plot is a **visualization used to summarize the distribution of one numerical variable** using:

- Minimum
- Q1 (25th percentile)
- Median (Q2)
- Q3 (75th percentile)
- Maximum
- Potential outliers

In EDA, a box plot is especially useful for understanding **spread and detecting potential outliers**.

## Explanation

A box plot summarizes the distribution of one numerical variable.

For example:

Hospital_Bill = [1000, 1200, 1500, 1800, 2000, 2200, 2500, 15000]

Most bills are between ₹1,000 and ₹2,500.

The value ₹15,000 is much higher than the other values.

Therefore, the box plot can help us identify ₹15,000 as a **potential outlier**.

The box represents the **middle 50% of the data**, which is the Interquartile Range (IQR).

IQR = Q3 - Q1

Potential outliers are commonly identified using:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Here, we are visualizing only the **Hospital_Bill** variable, so this is **univariate analysis**.

## Business Example

A hospital wants to analyze patient treatment bills.

Most patients may have bills between ₹1,000 and ₹2,500, while a few patients may have very high bills because they underwent expensive surgeries.

A box plot can quickly show those unusually high bills.

The hospital can then investigate whether the high values are:

- Genuine expensive treatments
- Special medical cases
- Billing errors
- Data-entry errors

## AI/ML Use

Box plots are useful during:

- Exploratory Data Analysis (EDA)
- Detecting potential outliers
- Understanding feature spread
- Comparing distributions
- Identifying skewness
- Data-quality investigation

For example, before training a model using Hospital_Bill as a feature, an ML Engineer can use a box plot to identify unusually large values.

However, an outlier should **not automatically be removed**.

The ML Engineer must first determine whether it is:

- A genuine observation
- A data-entry error
- A measurement error
- A special business case

## Finding

A box plot summarizes the **distribution of one numerical variable** and is especially useful for understanding the median, quartiles, spread, and potential outliers. In EDA, it provides a quick visual way to identify unusual observations that require further investigation.

````python

## Python Example

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Hospital_Bill": [1000, 1200, 1500, 1800, 2000, 2200, 2500, 15000]
})

sns.boxplot(y=df["Hospital_Bill"])

plt.ylabel("Hospital Bill")
plt.title("Hospital Bill Box Plot")

plt.show()
````

# 20. Percentage Distribution

## Definition

Percentage distribution shows **what percentage of the total observations belongs to each category**.

In EDA, it helps us understand the **proportion of each category**, not just the number of observations.

Formula:

Percentage = (Frequency / Total Observations) × 100


## Explanation

There are 6 patients in total.

Cardiology appears 3 times:

Percentage = (3 / 6) × 100 = 50%

Neurology appears 2 times:

Percentage = (2 / 6) × 100 = 33.33%

Orthopedics appears 1 time:

Percentage = (1 / 6) × 100 = 16.67%

Therefore:

Cardiology → 50%

Neurology → 33.33%

Orthopedics → 16.67%

The percentages add up to approximately 100%.

Here, we are analyzing only the **Department** variable, so this is **univariate analysis**.

## Business Example

A hospital wants to understand the proportion of patients visiting each department.

The analysis shows:

Cardiology → 50% of patients

Neurology → 33.33% of patients

Orthopedics → 16.67% of patients

This percentage information can help with:

- Staff allocation
- Resource planning
- Department capacity planning
- Patient service planning

## AI/ML Use

Percentage distribution is useful during:

- Exploratory Data Analysis (EDA)
- Understanding categorical features
- Comparing category proportions
- Detecting class imbalance
- Identifying dominant or rare categories
- Data preprocessing

For example, in a disease prediction dataset:

Disease = Diabetes → 90%
Disease = No Diabetes → 10%

This indicates that the target variable is **imbalanced**.

An ML Engineer should identify this during EDA because a highly imbalanced target can affect model training and evaluation.

## Finding

Percentage distribution shows the **proportion of each category within one variable**. In EDA, it helps us understand category dominance, rare categories, and class imbalance more clearly than raw frequency alone.

````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106],
    "Department": [
        "Cardiology",
        "Neurology",
        "Cardiology",
        "Orthopedics",
        "Cardiology",
        "Neurology"
    ]
})

percentage_distribution = (
    df["Department"].value_counts(normalize=True) * 100
)

print(percentage_distribution)
````

In EDA, a pie chart is appropriate when there are **few categories** and we want to show their percentage contribution.

### When NOT to Use a Pie Chart

## Business Example

A hospital wants to understand the percentage of patients visiting each department.

Diabetes → 90%

No Diabetes → 10%
