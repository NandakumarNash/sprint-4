# Class Distribution

## Concept

In classification, the **target variable** contains categories or classes that the Machine Learning model needs to predict.

**Class Distribution** means checking how the target values are distributed across different classes.

### Example – Hospital Dataset

Suppose the target variable is `Disease`:

* `Yes` → Patient has the disease
* `No` → Patient does not have the disease

If 7 patients have `No Disease` and 3 patients have `Disease`, the class distribution is:

* No Disease → 7 patients
* Disease → 3 patients

This helps us understand whether the target classes are **balanced or imbalanced**.

### Interpretation

There are:

* **7 patients → No Disease**
* **3 patients → Disease**

So, the target variable has **two classes: Yes and No**.

Understanding class distribution helps us identify whether the dataset has **class imbalance**, which can affect Machine Learning model performance.


````python

## Python Example
import pandas as pd

df = pd.DataFrame({
    "Disease": ["Yes", "No", "No", "Yes", "No",
                "No", "No", "Yes", "No", "No"]
})

class_distribution = df["Disease"].value_counts()

print(class_distribution)
````

# Class Counts

## Concept

**Class Counts** means finding the **number of observations belonging to each class** in the target variable.

It tells us how many records belong to each category.

### Example – Hospital Dataset

Suppose the target variable is `Disease`:

* `Yes` → Patient has the disease
* `No` → Patient does not have the disease

If there are 10 patients:

* No Disease → 7 patients
* Disease → 3 patients

The **class counts** are:

```text
No     7
Yes    3
```

### Interpretation

* `No` → **7 patients**
* `Yes` → **3 patients**

So, **Class Counts tell us the exact number of records in each target class**.


````python

import pandas as pd

df = pd.DataFrame({
    "Disease": ["Yes", "No", "No", "Yes", "No",
                "No", "No", "Yes", "No", "No"]
})

class_counts = df["Disease"].value_counts()

print(class_counts)
````

# Class Percentages

## Concept

**Class Percentages** means finding the **percentage of observations belonging to each class** in the target variable.

It helps us understand the proportion of each class and compare their sizes easily.

### Example – Hospital Dataset

Suppose the target variable is `Disease`:

* `No` → 7 patients
* `Yes` → 3 patients
* Total → 10 patients

Class percentages:

* No Disease → **70%**
* Disease → **30%**

### Interpretation

* `No` → **70% of patients**
* `Yes` → **30% of patients**

So, **Class Percentages tell us what percentage of the dataset belongs to each target class**.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Disease": ["Yes", "No", "No", "Yes", "No",
                "No", "No", "Yes", "No", "No"]
})

class_percentages = df["Disease"].value_counts(normalize=True) * 100

print(class_percentages)

````

# Class Imbalance

## Concept

**Class Imbalance** occurs when one target class has **many more observations than another class**.

In classification, this can cause the Machine Learning model to become biased toward the majority class.

### Example – Hospital Dataset

Suppose we have 100 patients:

* No Disease → 90 patients
* Disease → 10 patients

Here, the classes are highly imbalanced.

```text
No Disease → 90%
Disease    → 10%
```

If a model predicts **No Disease for every patient**, it will still achieve **90% accuracy**, but it completely fails to identify patients with the disease.


### Interpretation

* `No Disease` → **90%**
* `Disease` → **10%**

The dataset has **class imbalance** because one class is much larger than the other.

### Why is Class Imbalance Important?

Class imbalance can:

* Bias the model toward the majority class
* Make accuracy misleading
* Cause poor prediction of the minority class

Therefore, we should check class imbalance **before training a classification model**.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Disease": ["No"] * 90 + ["Yes"] * 10
})

class_counts = df["Disease"].value_counts()
class_percentages = df["Disease"].value_counts(normalize=True) * 100

print(class_counts)
print(class_percentages)
````

# Target Distribution

## Concept

For **regression problems**, the target variable contains **continuous numerical values**.

**Target Distribution** means understanding how the target values are spread across the dataset.

### Example – House Price Dataset

Suppose the target variable is `House_Price`.

```text
House_Price
250000
300000
280000
450000
320000
500000
```

We can analyze whether the prices are:

* Concentrated around a particular value
* Spread widely
* Skewed toward higher or lower values
* Containing unusual values or outliers

### Interpretation

The histogram shows how frequently different **house price ranges** occur.

For example, if most values are concentrated around ₹2.5–₹3.5 lakh and only a few houses have very high prices, the target distribution is **not evenly spread**.

Understanding the target distribution helps us decide whether the regression target has **skewness, unusual values, or outliers** before training a Machine Learning model.


````python

## Python Example

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "House_Price": [250000, 300000, 280000, 450000,
                    320000, 500000, 350000, 270000]
})

df["House_Price"].plot(kind="hist", bins=5)

plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.title("House Price Distribution")
plt.show()
````

# Mean

## Concept

**Mean** is the average value of the target variable.

It is calculated by adding all target values and dividing by the total number of values.

### Example – House Price Dataset

Suppose the house prices are:

```text
250000, 300000, 280000, 450000, 320000
```

Mean:

```text
(250000 + 300000 + 280000 + 450000 + 320000) / 5
= 320000
```

So, the **average house price is ₹3,20,000**.


### Interpretation

The mean tells us the **average value of the regression target**.

It is useful for understanding the **central tendency** of the target variable, but it can be strongly affected by **outliers**.


````python
## Python Example


import pandas as pd

df = pd.DataFrame({
    "House_Price": [250000, 300000, 280000, 450000, 320000]
})

mean_price = df["House_Price"].mean()

print(mean_price)
````

# Median

## Concept

**Median** is the **middle value** when the target values are arranged in ascending or descending order.

It is useful for understanding the center of the target distribution, especially when the data contains **outliers**.

### Example – House Price Dataset

Suppose the house prices are:

```text
250000, 300000, 280000, 450000, 320000
```

After sorting:

```text
250000, 280000, 300000, 320000, 450000
```

The middle value is:

```text
300000
```

So, the **median house price is ₹3,00,000**.


### Interpretation

The median tells us the **middle value of the target variable**.

Unlike the mean, the median is **less affected by extreme values or outliers**.


````python

## Python Example
import pandas as pd

df = pd.DataFrame({
    "House_Price": [250000, 300000, 280000, 450000, 320000]
})

median_price = df["House_Price"].median()

print(median_price)
````

# Variance

## Concept

**Variance** measures how far the target values are spread out from their mean.

* **Low variance** → values are close to the mean
* **High variance** → values are widely spread from the mean

### Example – House Price Dataset

Suppose the house prices are:

```text
250000, 300000, 280000, 320000, 300000
```

The values are relatively close to each other, so the variance will be relatively low.

If the prices were:

```text
100000, 200000, 300000, 700000, 1000000
```

the values are much more spread out, so the variance will be high.


### Interpretation

The variance tells us **how much the house prices vary around their mean**.

A higher variance means the target values have **greater spread**, while a lower variance means the values are **more concentrated around the mean**.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "House_Price": [250000, 300000, 280000, 320000, 300000]
})

variance = df["House_Price"].var()

print(variance)
````

# Standard Deviation

## Concept

**Standard Deviation** measures how much the target values typically differ from the mean.

It is the **square root of variance**.

* **Low standard deviation** → values are close to the mean
* **High standard deviation** → values are widely spread

### Example – House Price Dataset

Suppose the house prices are:

```text
250000, 300000, 280000, 320000, 300000
```

The prices are relatively close to each other, so the standard deviation is relatively low.

### Interpretation

The standard deviation is approximately **₹27,313**.

This means the house prices typically vary by around **₹27,313 from the mean price**.

Standard deviation is easier to interpret than variance because it is expressed in the **same unit as the target variable**.


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "House_Price": [250000, 300000, 280000, 320000, 300000]
})

std_price = df["House_Price"].std()

print(std_price)

````

# Skewness

## Concept

**Skewness** measures the **asymmetry of the target variable's distribution**.

It tells us whether the target values are balanced around the center or have a longer tail on one side.

* **Skewness ≈ 0** → approximately symmetric
* **Positive skewness** → tail extends toward higher values
* **Negative skewness** → tail extends toward lower values

### Example – House Price Dataset

Suppose most houses have prices between ₹2–₹4 lakh, but a few expensive houses cost ₹10 lakh.

```text id="6qg2v4"
250000, 280000, 300000, 320000, 400000, 1000000
```

Most values are relatively low, while one value is much higher.

This creates **positive skewness (right-skewed distribution)**.


### Interpretation

A **positive skewness value** indicates that the target distribution is **right-skewed**.

Understanding skewness is important because a highly skewed target can affect some Machine Learning models and may require appropriate data transformation.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "House_Price": [250000, 280000, 300000, 320000, 400000, 1000000]
})

skewness = df["House_Price"].skew()

print(skewness)
````

# Outliers

## Concept

**Outliers** are values that are **unusually far from most other values** in the target variable.

In regression, outliers in the target can strongly affect the **mean, variance, standard deviation, and Machine Learning model**.

### Example – House Price Dataset

Suppose most houses cost between ₹2 lakh and ₹5 lakh, but one house costs ₹50 lakh.

```text
250000, 280000, 300000, 320000, 450000, 5000000
```

Here, `5000000` is an **outlier** because it is much higher than the other house prices.

### Interpretation

If `5000000` appears in the output, it is detected as an **outlier**.

Before Machine Learning, we should investigate whether the outlier is:

* A genuine observation
* A data-entry error
* An unusual but valid case

Outliers should **not automatically be removed** without understanding why they occurred.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "House_Price": [250000, 280000, 300000, 320000, 450000, 5000000]
})

Q1 = df["House_Price"].quantile(0.25)
Q3 = df["House_Price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["House_Price"] < lower_limit) |
    (df["House_Price"] > upper_limit)
]

print(outliers)

````

# Why Target Variable Analysis Is Important

## Concept

The **target variable** is the value that a Machine Learning model tries to predict.

Before training a model, we should understand the target variable because its characteristics directly affect **model selection, preprocessing, evaluation, and performance**.

### For Classification

We should check:

* **Class Distribution** → How the classes are distributed
* **Class Counts** → Number of records in each class
* **Class Percentages** → Percentage of records in each class
* **Class Imbalance** → Whether one class is much larger than another

### For Regression

We should check:

* **Target Distribution** → How target values are spread
* **Mean** → Average target value
* **Median** → Middle target value
* **Variance** → Spread of target values
* **Standard Deviation** → Typical distance from the mean
* **Skewness** → Whether the distribution is asymmetric
* **Outliers** → Unusually high or low target values

### Example

For a hospital classification problem:

```text
Disease

No  → 90%
Yes → 10%
```

The target is highly imbalanced. A model could predict `No` for almost everyone and still achieve high accuracy, while performing poorly on patients with the disease.

For a house-price regression problem:

```text
Most prices → ₹2–₹5 lakh
Few prices → ₹50 lakh
```

The target contains extreme values and may be highly right-skewed.

### Why Analyze the Target Before Machine Learning?

Target analysis helps us:

* Understand what the model needs to predict
* Detect class imbalance
* Identify unusual target values
* Understand the target's spread and shape
* Choose suitable preprocessing techniques
* Select appropriate evaluation metrics
* Avoid misleading model performance

### Final Understanding

**Target Variable Analysis = Understand the value we are trying to predict before training the Machine Learning model.**

It is an important part of **EDA because the target variable is the final outcome that the model must learn to predict**.
